```python
%%manim -ql Gradients1

class Gradients1(Scene):
    def construct(self):
        gradients = [
            ["#159957", "#155799"], # 0
            ["#007991", "#78ffd6"],
            ["#56CCF2", "#2F80ED"],
            ["#F2994A", "#F2C94C"],
            ["#E44D26", "#F16529"],
            ["#30E8BF", "#FF8235"],
            ["#2E3192", "#1BFFFF"],
            ["#D4145A", "#FBB03B"],
            ["#009245", "#FCEE21"],
            ["#662D8C", "#ED1E79"], 
            ["#02AABD", "#00CDAC"], # 10
            ["#FF512F", "#DD2476"],
            ["#FF5F6D", "#FFC371"],
            ["#11998E", "#38EF7D"],
            ["#C6EA8D", "#FE90AF"],
            ["#FF61D2", "#FE9090"],
            ["#BFF098", "#6FD6FF"],
            ["#4E65FF", "#92EFFD"],
            ["#C33764", "#1D2671"],
            ["#34e89e", "#0f3443"],
            ["#6190E8", "#A7BFE8"], # 20
            ["#43C6AC", "#191654"],
            ["#3494E6", "#EC6EAD"],
            ["#67B26F", "#4ca2cd"],
            ["#ee0979", "#ff6a00"],
            ["#00c3ff", "#ffff1c"],
            ["#ff00cc", "#333399"],
            ["#3a6186", "#89253e"],
            ["#f79d00", "#64f38c"], 
            ["#0B486B", "#F56217"], 
            ["#2196f3", "#f44336"], # 30
            ["#FFFFFF", "#6284FF", "#FF0000"],
            ["#FCCF31", "#F55555"],
            ["#A0FE65", "#FA016D"],
            ["#FDD819", "#E80505"],
            ["#FFF720", "#3CD500"],
            ["#FD6585", "#0D25B9"],
            ["#69FF97", "#00E4FF"],
            ["#FFA8A8", "#FCFF00"], 
            ["#84fab0", "#8fd3f4"], 
            ["#43e97b", "#38f9d7"], # 40
            ["#fa709a", "#fee140"],
            ["#FEE140", "#FA709A"],
            ["#FF3CAC", "#784BA0", "#2B86C5"],
            ["#FAD961", "#F76B1C"]
        ]
        
        text = "This is gradient"

        for i in range(len(gradients)):
            copy = Tex(text + f" {i}")
            x = (i//15)*5 - 5
            y = 0.5*(i%15) - 3.5
            copy.shift(RIGHT*x + DOWN*y)
            copy.set_color_by_gradient(*gradients[i])
            self.add(copy)
            
```
