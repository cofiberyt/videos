```python
from manim import *
config.media_width = "40%"
config.verbosity = "WARNING"

config.frame_size = (1080, 1920) 
config.frame_width = 6
```

```python
%%manim S3

class S3(Scene):
    def construct(self):
        def pnorm(p):
            return lambda x,y : abs(x)**p + abs(y)**p - 1

        euclid = MathTex(r"\|x\|_2 = \left(\sum_{i=1}^n |x_i|^2 \right)^{1/2}")
        euclid.set_color(RED)
        euclid.move_to(UP*3.7)
        tax = MathTex(r"\|x\|_1 = \sum_{i=1}^n |x_i|")
        tax.set_color(BLUE)
        tax.move_to(DOWN*3.3)
        maxi = MathTex(r"\|x\|_\infty = \max_i |x_i|")
        maxi.set_color(TEAL)
        maxi.move_to(DOWN*4.5)
        
        normtex = MathTex(r"\|x\|_p", r"= \left(\sum_{i=1}^n |x_i|^p\right)^{1/p}")
        normtex.move_to(UP*3.7)
        normcon = MathTex(r"1 \le p < \infty")
        normcon.move_to(DOWN*3.3)
    
        axes = Axes(
            x_range = [-1.2, 1.2, 0.5],
            y_range = [-1.2, 1.2, 0.5],
            x_length = 5,
            y_length = 5,
            axis_config = {
                "numbers_to_include": np.arange(-1.0, 1.0 + 0.5, 0.5),
                "font_size": 24,
            },
            tips = False,
        )

        graph2 = axes.plot_implicit_curve(
            pnorm(2),
            color=RED
        )

        graph1 = axes.plot_implicit_curve(
            pnorm(1),
            color = BLUE
        )

        graphInfinity = axes.plot_implicit_curve(
            lambda x,y: max(abs(x),abs(y)) - 1,
            color = TEAL
        )

        graph5 = axes.plot_implicit_curve(
            pnorm(1.5),
            color = YELLOW
        )
        graph6 = axes.plot_implicit_curve(
            pnorm(4.5),
            color = ORANGE
        )
        graph7 = axes.plot_implicit_curve(
            pnorm(10),
            color = PINK
        )     
        #self.add(axes, euclid, tax, maxi)

        self.play(FadeIn(axes))
        self.play(Create(graph2), Write(euclid))
        self.play(Create(graph1), Write(tax))
        self.play(Create(graphInfinity), Write(maxi))
        self.wait(0.5)

        self.play(FadeOut(euclid, tax, maxi))
        self.play(Write(normtex), Write(normcon))
        

        self.play(Create(graph5))
        self.play(Create(graph6))
        self.play(Create(graph7))
        self.wait(2)

        #self.play(FadeOut(graph1, graph2, graphInfinity, graph5, graph6, graph7, axes, normtex, normcon))

"""
The sphere with respect to the euclidean norm is a circle. In the taxicab norm and the maximums norm its a square. These norms are all part of a larger 
class of norms called p norms. The sphere with respect to these norms fill the gaps in our picture.
"""
```

```python
%%manim S4

class S4(Scene):
    def construct(self):
        lpnorm = MathTex(r"\|f\|_p = \left(\int_X |f|^p\,d\mu\right)^{1/p}")
        lpspace = Tex(r"$L^p$-Spaces", font_size=100)
        lpspace.set_color_by_gradient("#4E65FF", "#92EFFD")
        holder = MathTex(r"\int_X |fg|\,d\mu \le \|f\|_p\|g\|_q")
        mink = MathTex(r"\|f+g\|_p \le \|f\|_p + \|g\|_p")

        lpnorm.move_to(UP*3.5)
        lpspace.move_to(UP)
        holder.move_to(DOWN*1.5)
        mink.move_to(DOWN*3)

        self.play(FadeIn(lpnorm))
        self.play(Write(lpspace))
        self.wait()
        self.play(FadeIn(holder))
        self.play(FadeIn(mink))
        self.wait()

"""
We can generalize these norms and obtain L-p norms and the corresponding L-p spaces. These are used all over mathematics. 
In those spaces, we can for instance prove the Hölder inequality and the Minkowski inequality. 

Watch the full video if you are interested.
"""
```

```python
%%manim Young

class Young(Scene):
    def construct(self):
        funcdef = Tex(r"$f$ continuous strictly \\increasing function")
        funcdef.move_to(UP*4.3)

        axes = Axes(
            x_range = [0, 5, 1],
            y_range = [0, 5, 1],
            x_length = 5,
            y_length = 5,
            tips = False
        )
        axes.move_to(UP*1.2)

        func = lambda x: np.exp(-x/5) * np.sin(np.pi*x/3) + x

        a = 2
        b = 3.5
        a1 = 4
        b1 = 2.5

        curve = axes.plot(
            func,
            color=BLUE
        )
        curve.z_index = 1
        curve2 = axes.plot(
            lambda x : func(b)
        )
        curve3 = axes.plot(
            lambda x : func(b1)
        )

        area1 = axes.get_area(curve, x_range = [0,a], color = BLUE, opacity = 0.5)
        area12 = axes.get_area(curve, x_range = [0,a1], color = BLUE, opacity = 0.5)
        area2 = axes.get_area(curve, x_range=[0, b], bounded_graph=curve2, color=RED, opacity = 0.5)
        area22 = axes.get_area(curve, x_range=[0, b1], bounded_graph=curve3, color=RED, opacity = 0.5)
        
        pa = Dot(radius = 0.06)
        labela = MathTex("a")
        pa.move_to(axes.c2p(a,0))
        labela.next_to(pa, DOWN)
        pb = Dot(radius = 0.06)
        labelb = MathTex("b")
        pb.move_to(axes.c2p(0,func(b)))
        labelb.next_to(pb, LEFT)

        pa1 = Dot(radius = 0.06)
        labela1 = MathTex("a")
        pa1.move_to(axes.c2p(a1,0))
        labela1.next_to(pa1, DOWN)
        pb1 = Dot(radius = 0.06)
        labelb1 = MathTex("b")
        pb1.move_to(axes.c2p(0,func(b1)))
        labelb1.next_to(pb1, LEFT)
        
        rect = Polygon(axes.c2p(0,0), axes.c2p(a,0), axes.c2p(a,func(b)), axes.c2p(0,func(b)), color = ORANGE)
        rect1 = Polygon(axes.c2p(0,0), axes.c2p(a1,0), axes.c2p(a1,func(b)), axes.c2p(0,func(b)), color = ORANGE)
        rect2 = Polygon(axes.c2p(0,0), axes.c2p(a1,0), axes.c2p(a1,func(b1)), axes.c2p(0,func(b1)), color = ORANGE)
        rect.z_index = 2

        self.play(FadeIn(axes))
        self.play(Write(funcdef), Create(curve))
        self.wait()
        self.play(FadeIn(area1, pa, labela))
        self.wait()
        self.play(FadeIn(area2, pb, labelb))
        self.wait()
        self.play(Create(rect))
        self.wait(2)

        self.play(Transform(pa, pa1), Transform(labela, labela1), Transform(rect, rect1), ReplacementTransform(area1, area12))
        self.wait()
        self.play(Transform(pb, pb1), Transform(labelb, labelb1), Transform(rect, rect2), ReplacementTransform(area2, area22))
        self.wait()

        graph = VGroup(pa, pb, labela, labelb, rect, area12, area22, axes, curve)

        inequ = MathTex(r"ab \le ", r"\int_0^a f(x)\,dx", "+", r"\int_0^b f^{-1}(x)\,dx", font_size=35)
        inequ[1].set_color(BLUE)
        inequ[3].set_color(RED)
        inequ.move_to(DOWN*2.5)

        self.play(Write(inequ))
        self.wait()
        self.play(FadeOut(graph, funcdef), inequ.animate.shift(UP*6.5))
        self.wait()

        young01 = MathTex(r"p,q \in (1,\infty), \frac{1}{p} + \frac{1}{q} = 1")
        young1 = MathTex(r"f(x) = x^{p-1}", r",", r"f^{-1}(x) = x^{q-1}", font_size=40)
        young1[0].set_color(BLUE)
        young1[2].set_color(RED)
        young01.move_to(UP*2.5)
        young1.next_to(young01, DOWN)


        self.play(FadeIn(young01, young1))
        young2 = MathTex(r"ab \le", r"\frac{a^p}{p}", r"+", r"\frac{b^q}{q}")
        young3 = Tex(r"Young's inequality")
        young3.next_to(young2, DOWN)
        young3.set_color_by_gradient("#4E65FF", "#92EFFD")
        self.wait()
        self.play(Write(young2))
        self.play(Write(young3))
        self.wait()

"""
Suppose you have a continuous strictly increasing function. When you look at the area under the curve and over the curve up to given points, you see
that their combined area is always greater than the area of this rectangle. We can express this with integrals like this.
When we take this particular function as f, we get Young's inequality.
"""
```
