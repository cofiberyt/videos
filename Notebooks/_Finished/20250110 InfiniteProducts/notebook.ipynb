```python
%%manim -qm S1

class S1(Scene):
    def construct(self):
        basel = Tex(r"Basel problem: ", r"$\sum_{n=1}^\infty \frac{1}{n^2} = \frac{\pi^2}{6}$")
        basel[0].set_color_by_gradient(*gradient_colors)
        sine = MathTex(r"\sin(x) = x \prod_{n=1}^\infty \left(1-\frac{x^2}{\pi^2n^2}\right)")
        basel.move_to(UP*3)
        sine.move_to(UP*1.5)

        axes = Axes(
            x_range=[-4*PI, 4*PI, PI],
            y_range=[-1.5, 1.5, 0.5],
            x_length=10,
            y_length=4,
            tips=False
        )
        axes.move_to(DOWN*1.5)

        sine_graph = axes.plot(
            lambda x: np.sin(x),
            color=BLUE,
            x_range=[-4*PI, 4*PI]
        )

        self.play(Write(basel))
        self.wait()
        self.play(Write(sine))

        self.play(FadeIn(axes))
        self.play(Create(sine_graph))

        self.wait()

        self.play(FadeOut(basel, sine, axes, sine_graph))

        poly = MathTex(r"\text{Polynomial: }", r"p(x) = \sum_{n=0}^d p_nx^n")
        poly[0].set_color_by_gradient(*gradient_colors)
        roots = Tex(r"Zeros ($\ne 0$): ", r"$a_1, a_2, \dots, a_m$")
        fact = MathTex(r"p(x) = p_0z^{d-m}\prod_{n=1}^d \left(1 - \frac{x}{a_n}\right)")
        comfunc = MathTex(r"p:\mathbb{C} \to \mathbb{C}")

        poly.move_to(UP*2)
        fact.move_to(DOWN*2)

        self.play(Write(poly))
        self.wait()
        self.play(Write(roots))
        self.wait()
        self.play(Write(fact))
        self.wait()
        self.play(Transform(roots, comfunc))
        self.wait()
        self.play(poly.animate.shift(UP*0.5 + LEFT*3.5), fact.animate.shift(UP*4.5 + RIGHT*3.5), FadeOut(roots))

        inf = MathTex(r"\sin(x) = \sum_{n=0}^\infty \frac{(-1)^n}{(2n+1)!}x^{2n+1}", r"\text{ 'Infinite degree polynomial'}")
        inf[1].set_color_by_gradient(*gradient_colors)
        #inf.move_to(UP*0.5)
        sine2 = MathTex(r"\sin(x) = x \prod_{n=1}^\infty \left(1-\frac{x^2}{\pi^2n^2}\right)")
        sine2.move_to(DOWN*2)

        self.play(Write(inf))
        self.wait()
        self.play(Write(sine2))
        self.wait()
        self.play(FadeOut(poly, fact, inf, sine2))
```

```python
%%manim -qm S7

class S7(Scene):
    def elem(self, n):
        return lambda x : (1-x)* np.exp(sum([x**i / i for i in range(1, n+1)]))
    
    def construct(self):
        factors = MathTex(r"\prod_{n=1}^\infty \left(1-\frac{z}{a_n} \right) \exp\left(g_n\left(\frac{z}{a_n}\right)\right)")
        factors.move_to(UP*3)
        limit = MathTex(r"\frac{z}{a_n} \xrightarrow{n\to\infty} 0")
        limit.move_to(UP*1.5)
        res1 = Tex(r"Choose $g_n$ such that $\left(1-\frac{z}{a_n} \right) \exp\left(g_n\left(\frac{z}{a_n}\right)\right)$ \\is close to $1$ for $|\frac{z}{a_n}|<1$")
        res2 = MathTex(r"\log(1-z) = -\sum_{k=1}^\infty \frac{z^k}{k} \text{ if } |z|<1")
        res2.move_to(DOWN*1.5)
        res3 = MathTex(r"E_n(z) = (1-z)\exp \left(\sum_{k=1}^n \frac{z^k}{k}\right)")
        res3.move_to(DOWN*3)

        self.play(Write(factors))
        self.wait()
        self.play(Write(limit))
        self.wait()
        self.play(Write(res1))
        self.wait()
        self.play(Write(res2))
        self.wait()
        self.play(Write(res3))
        self.wait()

        cantitle = Tex(r"Elementary factors")
        cantitle.move_to(UP*3.3)
        cantitle.set_color_by_gradient(*gradient_colors)
        self.play(FadeOut(factors, limit, res1, res2), res3.animate.shift(UP*5))
        self.play(Write(cantitle))
        self.wait()

        
        axes = Axes(
            x_range=[-1, 1, 0.5], 
            y_range=[0, 2, 0.5],
            x_length=8,
            y_length=4,
            tips=False,
            x_axis_config = {"numbers_to_include": [-1,-0.5,0,0.5,1]},
            y_axis_config = {"numbers_to_include": [0,0.5,1,1.5,2]}
        )
        axes.next_to(res3, DOWN)
        self.play(FadeIn(axes))

        colors = [BLUE, GREEN, RED, ORANGE, PURPLE, TEAL, YELLOW, MAROON, PINK, GOLD, GRAY, WHITE]
        graphs = VGroup()
        for i in range(5):
            graph = axes.plot(self.elem(i), color=colors[i])
            self.play(Create(graph))
            self.wait(0.2)
            graphs.add(graph)

        self.wait()

        tex51 = MathTex(r"|1-E_n(z)| \le |z|^{n+1}")
        tex52 = Tex(r"if $|z|\le1$")
        tex52.next_to(tex51, DOWN)
        tex5 = VGroup(tex51, tex52)
        tex5.move_to(RIGHT*4.3 + DOWN*0.7)
        self.play(VGroup(axes, graphs).animate.shift(LEFT*2.5), Write(tex5))
        self.wait()
        self.play(FadeOut(cantitle, axes, graphs, tex5, res3))
```

```python
%%manim -qm S11

class S11(Scene):
    def construct(self):
        axes = Axes(
            x_range=[-4*PI, 4*PI, PI],
            y_range=[-1.5, 1.5, 0.5],
            x_length=10,
            y_length=4,
            tips=False
        )

        sine_graph = axes.plot(
            lambda x: np.sin(x),
            color=BLUE,
            x_range=[-4*PI, 4*PI]
        )
        
        sine = MathTex(r"\sin(\pi z)")
        zeros = Tex(r"simple zeros at $z\in \mathbb{Z}$")
        func1 = MathTex(r"\sin(\pi z) = e^{c_1z+c_2}z\prod_{n \in \mathbb{Z}\setminus\{0\}} E_1\left(\frac{z}{n}\right)")
        func2 = MathTex(r"\sin(\pi z) = e^{c_1z+c_2}z\prod_{n \in \mathbb{Z}\setminus\{0\}} \left(1-\frac{z}{n}\right)e^{z/n}")

        sine.move_to(UP*3)
        zeros.move_to(UP*2)

        self.play(FadeIn(axes))
        self.play(Create(sine_graph))
        self.wait()
        plot = VGroup(axes, sine_graph)
        self.play(plot.animate.shift(DOWN*1.7), Write(sine))
        self.wait()
        self.play(Write(zeros))
        self.wait()

        inequ = MathTex(r"|\sin(\pi z)| = \frac{1}{2}|e^{\pi i z} - e^{-\pi i z}| \le e^{\pi |z|}")
        order = Tex(r"$\leadsto$ order of growth $1$")
        inequ.move_to(UP*2)
        order.next_to(inequ, DOWN)
        self.play(sine.animate.shift(LEFT*3), zeros.animate.shift(RIGHT*3 + UP))
        self.play(Write(inequ))
        self.wait()
        self.play(Write(order))
        self.wait()
        self.play(FadeOut(inequ), order.animate.shift(UP*1))
        self.wait()
        self.play(FadeOut(plot))
        self.play(Write(func1))
        self.wait()
        self.play(Transform(func1, func2))
        self.wait()

        con1 = MathTex(r"\lim_{z\to 0} \frac{\sin(\pi z)}{\pi z} = 1", r"\Rightarrow e^{c_2} = \pi")
        con1.move_to(DOWN*2)
        func3 = MathTex(r"\sin(\pi z) = e^{c_1z}\pi z\prod_{n \in \mathbb{Z}\setminus\{0\}} \left(1-\frac{z}{n}\right)e^{z/n}")

        self.play(Write(con1[0]))
        self.wait()
        self.play(Write(con1[1]))
        self.wait()
        self.play(Transform(func1, func3), FadeOut(con1))
        self.wait()

        con2 = MathTex(r"\sin(\pi(-z)) = -\sin(\pi z)")
        con22 = MathTex(r"\Rightarrow e^{c_1(-z)} \pi (-z) = -e^{c_1 z} \pi z")
        con23 = MathTex(r"\Rightarrow e^{c_1(-z)} = e^{c_1 z}")
        con24 = MathTex(r"\Rightarrow c_1 = 0")
        func4 = MathTex(r"\sin(\pi z) = \pi z\prod_{n \in \mathbb{Z}\setminus\{0\}} \left(1-\frac{z}{n}\right)e^{z/n}")
        con2.move_to(DOWN*2)
        con22.move_to(DOWN*3)
        con23.move_to(DOWN*3)
        con24.move_to(DOWN*3)

        self.play(Write(con2))
        self.wait()
        self.play(Write(con22))
        self.wait()
        self.play(Transform(con22, con23))
        self.wait()
        self.play(Transform(con22, con24))
        self.wait()
        self.play(Transform(func1, func4), FadeOut(con2, con22))
        self.wait()

        tex76 = MathTex(r"\left(1-\frac{z}{n}\right )e^{z/n}\left(1-\frac{z}{-n}\right )e^{z/(-n)} = \left(1 - \frac{z^2}{n^2}\right)")
        func5 = MathTex(r"\sin(\pi z) = \pi z\prod_{n \in \mathbb{Z}\setminus\{0\}} \left(1-\frac{z^2}{n^2}\right)")
        tex76.move_to(DOWN*2)

        self.play(FadeIn(tex76))
        self.wait()
        self.play(Transform(func1, func5))
        self.play(FadeOut(tex76))
        self.wait()
        self.play(FadeOut(func1, sine, order, zeros))
```

```python
%%manim -qm S16

class S16(Scene):
    def construct(self):
        title = Tex(r"Jensen's formula")
        title.set_color_by_gradient(*gradient_colors)
        title.move_to(UP*2)
        con = MathTex(r"\log |f(0)| + \sum_{n=1}^{n(r)} \log \left(\frac{r}{|a_n|}\right) = \frac{1}{2\pi}\int_{-\pi}^\pi \log |f(re^{i\theta})|\,d\theta")
        con.next_to(title, DOWN)

        res1 = MathTex(r"\sup_{0<r<1}\int_{-\pi}^\pi \log |f(re^{i\theta})|\,d\theta \le C")
        res1.move_to(DOWN*2)
        
        self.play(FadeIn(title, con))
        self.wait()
        self.play(Write(res1))
        self.wait()

        con2 = MathTex(r"|f(0)|\prod_{n=1}^{k} \frac{r}{|a_n|} \le e^C \text{ for all } r \text{ with } n(r) > k")
        con2.next_to(title, DOWN)

        self.play(FadeOut(title), Transform(con, con2))
        self.wait()

        con3 = MathTex(r"e^{-C}|f(0)| \le \prod_{n=1}^k |a_n|")
        con3.next_to(title, DOWN)
        self.play(Transform(con, con3))
        self.wait()

        con4 = MathTex(r"0 < e^{-C}|f(0)| \le \prod_{n=1}^\infty |a_n|")
        con4.next_to(title, DOWN)
        self.play(Transform(con, con4))
        self.wait()

        bla1 = Tex(r"Blaschke condition")
        bla1.set_color_by_gradient(*gradient_colors)
        bla2 = MathTex(r"\sum_{n=1}^\infty (1 - |a_n|) < \infty")
        bla2.next_to(bla1, DOWN)
        bla = VGroup(bla1, bla2)
        bla.move_to(ORIGIN)

        self.play(FadeOut(res1, con), FadeIn(bla))
        self.wait()
        self.play(FadeOut(bla))
```

```python
%%manim -qm S19

class S19(Scene):
    def construct(self):
        def1 = Tex(r"$f:\mathbb{D} \to \mathbb{C}$ holomorphic")
        def2 = MathTex(r"f_r(e^{i\theta}) = f(re^{i\theta}) \quad (0 < r < 1)")
        def3 = MathTex(r"\|f\|_p = \sup_{0<r<1} \left( \frac{1}{2\pi}\int_{-\pi}^\pi |f_r(e^{i\theta})|^p\,d\theta \right)^{1/p} \quad (0<p<\infty)")
        def4 = MathTex(r"\|f\|_\infty = \sup_{0<r<1} \sup_{\theta} |f_r(e^{i\theta})|")
        def5 = MathTex(r"\text{Hardy spaces }", r"H^p = \{f:\mathbb{D} \to \mathbb{C} \text{ holomorphic } \mid \|f\|_p < \infty\}")
        def5[0].set_color_by_gradient(*gradient_colors)
        def6 = MathTex(r"p < q \Rightarrow H^\infty \subset H^q \subset H^p")

        def1.move_to(UP*3)
        def2.move_to(UP*2)
        def3.move_to(UP*0.5)
        def4.move_to(DOWN*1)
        def5.move_to(DOWN*2.5)
        def6.next_to(def5, DOWN)

        self.play(FadeIn(def1))
        self.wait()
        self.play(Write(def2))
        self.wait()
        self.play(FadeIn(def3, def4))
        self.wait()
        self.play(FadeIn(def5))
        self.wait()
        self.play(FadeIn(def6))
        self.wait()
        self.play(FadeOut(def1, def2, def3, def4, def6), def5.animate.shift(UP*5.5))

        rad1 = MathTex(r"f \in H^p")
        rad2 = MathTex(r"f^*(e^{i\theta}) = \lim_{r\to 1^-} f_r(e^{i\theta}) \quad (a.e.)")
        rad3 = MathTex(r"f,g \in H^p")
        rad4 = Tex(r"If $f^* = g^*$ on a subset of \\positive Lebesgue measure, then $f = g$.")
        
        rad1.move_to(UP*1.6)
        rad2.next_to(rad1, DOWN)
        rad3.move_to(DOWN)
        rad4.next_to(rad3, DOWN)

        self.play(FadeIn(rad1))
        self.play(Write(rad2))
        self.wait()
        self.play(FadeIn(rad3))
        self.play(Write(rad4))
        self.wait()
        self.play(FadeOut(def5, rad1, rad2, rad3, rad4))
```
