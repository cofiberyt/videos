```python
%%manim -qh MeasureTheory

class MeasureTheory(Scene):
    def construct(self):
        gradient_colors = ["#FF512F", "#DD2476"]
        
        title = Tex("Measure Theory", font_size=165)
        title.set_color_by_gradient(*gradient_colors)
        self.add(title)
        
```

```python
%%manim -qh RadonNikodym

class RadonNikodym(Scene):
    def construct(self):
        gradient_colors = ["#009245", "#FCEE21"]
        
        title = Tex(r"Generalizing the Derivative", font_size=100)
        title.move_to(UP*2.5)
        title.set_color_by_gradient(*gradient_colors)
        self.add(title)

        normal = MathTex(r"\frac{df}{dx}", font_size = 100)
        normal.move_to(LEFT*3 + DOWN*0.5)
        arrow = MathTex(r"\longrightarrow", font_size = 100)
        arrow.move_to(DOWN*0.5)
        radon = MathTex(r"\frac{d\nu}{d\mu}", font_size = 100)
        radon.move_to(RIGHT*3 + DOWN*0.5)

        self.add(normal, radon, arrow)
```

```python
%%manim -ql AlgebraicNumberTheory

class AlgebraicNumberTheory(Scene):
    def construct(self):
        gradient_colors = ["#f79d00", "#64f38c"]

        title1 = Tex("Introduction to", font_size=110)
        title1.move_to(UP*2)
        title2 = Tex("Algebraic Number Theory", font_size=110)
        title2.move_to(DOWN)
        title2.set_color_by_gradient(*gradient_colors)

        self.add(title1, title2)
```

```python
%%manim -qh TauFunction

class TauFunction(Scene):
    def construct(self):
        gradient_colors = ["#009245", "#FCEE21"]
        
        tex = Tex(r"The Local-Global Principle", font_size=105)
        tex.set_color_by_gradient(*gradient_colors)
        tex.move_to(UP*2.5)

        self.add(tex)
        
```

```python
%%manim -ql EllipticCurves

class EllipticCurves(Scene):
    def construct(self):
        gradient_colors = ["#009245", "#FCEE21"]
        
        tex = Tex(r"Elliptic Curves", font_size=105)
        tex.set_color_by_gradient(*gradient_colors)
        tex.move_to(UP*3)

        a, b = -1, 1
        graph = ImplicitFunction(
            lambda x, y: y ** 2 - (x**3 + a*x + b),
            color=YELLOW,
            x_range=[-2, 2], y_range=[-2,2]
        )

        ax = Axes(
            x_range=[-2, 2], 
            y_range=[-2,2],
            x_length=4,
            y_length=4,
            tips=False,
        )
        ax.move_to(DOWN*0.5 + LEFT*3)
        a = ax.plot_implicit_curve(
            lambda x, y: y ** 2 - (x**3 + a*x + b),
            color=YELLOW
        )S
        
        self.add(a, ax, tex)
```
