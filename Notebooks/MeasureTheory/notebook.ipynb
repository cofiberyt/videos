# Part I - Constructions of measures

```python
%%manim -qm S1

class S1(Scene):
    def construct(self):
        title = Tex("Finite Additivity", font_size=60).to_edge(UP)
        tex = MathTex(r"E_k \text{ disjoint } \Rightarrow \operatorname{Area}(\bigcup_{k=1}^n E_k) = \sum_{k=1}^n \operatorname{Area}(E_k) ")
        tex.next_to(title, DOWN)
        
        svg = SVGMobject("FiniteAdditivity.svg", height=5).to_edge(DOWN, buff=0.2)
        subs = svg.submobjects
        
        ini = subs[0]
        parts = VGroup(*subs[1:])
        self.play(FadeIn(ini))
        self.wait()
        self.play(FadeIn(parts))
        self.play(parts[1].animate.shift(RIGHT*0.2+UP*0.1), parts[3].animate.shift(LEFT*0.2+UP*0.1), parts[2].animate.shift(DOWN*0.2))
        self.wait()
        self.play(FadeIn(tex))
        self.wait()
        self.play(Write(title))
        self.wait()

        tex2 = MathTex(r"E_k \text{ disjoint } \Rightarrow \operatorname{Area}(\bigcup_{k=1}^\infty E_k) \stackrel{?}{=} \sum_{k=1}^\infty \operatorname{Area}(E_k) ")
        tex2.next_to(title, DOWN)
        
        width = 4.0
        height = 4.0
        center = DOWN*1.5

        start = Rectangle(width=width, height=height).move_to(center)
        
        self.play(FadeIn(start), FadeOut(svg))
        self.wait()

        last = Rectangle(width=width, height=height).next_to(start, DOWN, buff=0)
        rects = VGroup()
        rects.add(start)
        
        for i in range(6):
            height /= 2
            last = Rectangle(width = width, height=height, fill_color=BLUE, fill_opacity=0.3).next_to(last, UP, aligned_edge=RIGHT, buff=0)
            self.play(FadeIn(last))
            rects.add(last)
            
            width /= 2
            rect2 = Rectangle(width=width, height=height, fill_color=BLUE, fill_opacity=0.3).next_to(last, UP, aligned_edge=LEFT, buff=0)
            self.play(FadeIn(rect2))
            rects.add(rect2)

        self.play(FadeOut(title), Transform(tex, tex2))
        self.wait()
        title2 = Tex(r"$\sigma$-Additivity", font_size=60).to_edge(UP)
        self.play(Write(title2))
        self.wait()
        self.play(FadeOut(tex), rects.animate.move_to(RIGHT*3.75), title2.animate.move_to(RIGHT*3.75+UP*3))
        parts.move_to(LEFT*3.75)
        title.move_to(LEFT*3.75 + UP*3)
        line = Line(start=UP*3.5, end=DOWN*3.5)
        self.play(FadeIn(parts, title, line))
        self.wait()

        fin1 = Tex(r"Content", font_size=60).move_to(LEFT*3.75 + DOWN*3.25)
        sig1 = Tex(r"Measure", font_size=60).move_to(RIGHT*3.75 + DOWN*3.25)

        self.play(Write(fin1))
        self.wait()
        self.play(Write(sig1))
        self.wait()
        self.play(sig1.animate.scale(1.5))
        self.wait()
        self.play(FadeOut(parts, title, line, rects, title2, fin1, sig1))

        
```

# Part II - Measurable Functions

```python
%%manim -qm LusinTheorem

def char(a,b):
    return lambda x: 1 if a <= x <= b else 0

def approChar(a, b, eps):
    def f(x):
        if x <= a-eps or x >= b+eps:
            return 0
        elif a + eps <= x and b - eps >= x:
            return 1
        elif a - eps < x < a + eps:
            return (x - a + eps)/(2*eps)
        elif b - eps < x < b + eps:
            return (b + eps - x)/(2*eps)
    return f
        
class LusinTheorem(Scene):
    def construct(self):
        grid = Axes(
            x_range=[0, 1, 0.1],  # step size determines num_decimal_places.
            y_range=[0, 1, 0.1],
            x_length=5,
            y_length=5,
            tips=False,
        )
        grid2 = grid.copy().shift(LEFT*3)
        grid.shift(RIGHT*3)
        self.add(grid, grid.plot(approChar(0.3, 0.7, 0.1), color=RED, x_range=[0, 1, 0.001]))
        self.add(grid2, grid2.plot(char(0.3, 0.7), color=BLUE, x_range=[0,1,0.001], discontinuities=[0.3,0.7], dt=0.001))
        self.wait()
        
```
