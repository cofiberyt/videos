```python
%%manim -qh MorletWavelet

class MorletWavelet(Scene):
    def construct(self):
        # Parameters for the Morlet wavelet
        omega_0 = 10
        sigma = 0.5

        circ = Circle(radius=3.5, color=GREY_E, stroke_width=15)

        # Define the Morlet wavelet function
        def morlet_wavelet(t):
            return np.pi**(-1/4) * np.exp(1j * omega_0 * t) * np.exp(-t**2 / (2 * sigma**2))

        # Create the graph of the Morlet wavelet (real part)
        wavelet = FunctionGraph(lambda t: np.real(morlet_wavelet(t)), x_range=[-1, 1])
        wavelet.set_color(cols1)
        wavelet.set_stroke(width=45)
        wavelet.scale(3.1)

        every = VGroup(wavelet, circ)
        self.add(every)
```

```python
def hexarot(mult):
    rot_angle = mult*PI/3

    poly = RegularPolygon(n=6)
    poly.set_stroke(color=cols3, width=8)
    circ = Circle(radius=1.2)

    if mult == 0:
        angletex = MathTex(r"0^{\circ}").set_color(BLUE)
        rot_poly = VGroup(poly.copy(), angletex)
        return rot_poly
    if mult == 1:
        angletex = MathTex(r"60^{\circ}").set_color(BLUE)
        arc1 = ArcBetweenPoints(start=circ.point_at_angle(0), end=circ.point_at_angle(rot_angle), radius=1.58)
        arc1.add_tip()
        arc2 = ArcBetweenPoints(start=circ.point_at_angle(PI), end=circ.point_at_angle(PI+rot_angle), radius=1.58)
        arc2.add_tip()
        rot_poly = VGroup(poly.copy(), angletex, arc1, arc2)
        return rot_poly
    if mult == 2:
        angletex = MathTex(r"120^{\circ}").set_color(BLUE)
        arc1 = ArcBetweenPoints(start=circ.point_at_angle(0), end=circ.point_at_angle(rot_angle), radius=1.32)
        arc1.add_tip()
        arc2 = ArcBetweenPoints(start=circ.point_at_angle(PI), end=circ.point_at_angle(PI+rot_angle), radius=1.32)
        arc2.add_tip()
        rot_poly = VGroup(poly.copy(), angletex, arc1, arc2)
        return rot_poly
    if mult == 3:
        angletex = MathTex(r"180^{\circ}").set_color(BLUE)
        arc1 = ArcBetweenPoints(start=circ.point_at_angle(0), end=circ.point_at_angle(rot_angle), radius=1.213)
        arc1.add_tip()
        arc2 = ArcBetweenPoints(start=circ.point_at_angle(PI), end=circ.point_at_angle(PI+rot_angle), radius=1.213)
        arc2.add_tip()
        rot_poly = VGroup(poly.copy(), angletex, arc1, arc2)
        return rot_poly
    if mult == -1:
        angletex = MathTex(r"-60^{\circ}").set_color(BLUE)
        arc1 = ArcBetweenPoints(end=circ.point_at_angle(0), start=circ.point_at_angle(rot_angle), radius=1.58)
        arc1.add_tip(at_start=True)
        arc2 = ArcBetweenPoints(end=circ.point_at_angle(PI), start=circ.point_at_angle(PI+rot_angle), radius=1.58)
        arc2.add_tip(at_start=True)
        rot_poly = VGroup(poly.copy(), angletex, arc1, arc2)
        return rot_poly
    if mult == -2:
        angletex = MathTex(r"-120^{\circ}").set_color(BLUE)
        arc1 = ArcBetweenPoints(end=circ.point_at_angle(0), start=circ.point_at_angle(rot_angle), radius=1.32)
        arc1.add_tip(at_start=True)
        arc2 = ArcBetweenPoints(end=circ.point_at_angle(PI), start=circ.point_at_angle(PI+rot_angle), radius=1.32)
        arc2.add_tip(at_start=True)
        rot_poly = VGroup(poly.copy(), angletex, arc1, arc2)
        return rot_poly
    if mult == -3:
        angletex = MathTex(r"-180^{\circ}").set_color(BLUE)
        arc1 = ArcBetweenPoints(end=circ.point_at_angle(0), start=circ.point_at_angle(rot_angle), radius=1.213)
        arc1.add_tip(at_start=True)
        arc2 = ArcBetweenPoints(end=circ.point_at_angle(PI), start=circ.point_at_angle(PI+rot_angle), radius=1.213)
        arc2.add_tip(at_start=True)
        rot_poly = VGroup(poly.copy(), angletex, arc1, arc2)
        return rot_poly

def hexaref(mult):
    ref_angle = mult*PI/6
    
    poly = RegularPolygon(n=6)
    poly.set_stroke(color=cols3, width=8)
    
    axis = DashedLine(1.5 * UP, 1.5 * DOWN, dash_length=0.2, dashed_ratio=0.75).rotate(ref_angle).set_color(BLUE)
    ref_poly = VGroup(poly.copy(), axis)
    return ref_poly

def refmatrix(theta):
    return np.array([
        [-np.cos(2*theta), -np.sin(2*theta), 0],
        [-np.sin(2*theta), np.cos(2*theta), 0],
        [0, 0, 1]
    ])
```

```python
%%manim -qh GroupEx3

class GroupEx3(Scene):
    def construct(self):
        title = Tex(r"Symmetric group $S_n$", font_size=110).set_color_by_gradient(*cols1)
        title.move_to(UP*1.5)
        perm = Tex(r"Permutations of $n$ elements\\ under composition", font_size=90)
        perm.move_to(DOWN*0.5)

        self.play(FadeIn(title))
        self.wait()
        self.play(FadeIn(perm))
        self.wait()
        self.play(FadeOut(title, perm))

        num_elements = 9
        circle_radius = 3.2
        arrow_stroke_width = 4
        tip_size = 0.2

        # Create numbered elements
        angles = np.linspace(0, 2*PI, num_elements, endpoint=False)
        dots = VGroup(*[
            VGroup(
                Dot(radius=0.12),
                Text(str(i), font_size=24)
            ).arrange(UP, buff=0.15).move_to(circle_radius * np.array([np.cos(a), np.sin(a), 0]))
            for i, a in enumerate(angles)
        ])
        
        # Permutation definition with proper structure
        perm_cycles = [
            ([0, 1, 2, 3, 4], RED, 1.2),    # 5-cycle
            ([5, 6, 7], GREEN, 0.6),       # 3-cycle
            ([8], BLUE, 0)                 # Fixed point (angle unused)
        ]

        # Arrow creation function
        def create_arrow(start_idx, end_idx, color, arc_angle):
            return CurvedArrow(
                dots[start_idx][0].get_center(),
                dots[end_idx][0].get_center(),
                stroke_width=arrow_stroke_width,
                tip_length=tip_size,
                angle=arc_angle,
                arc_center=ORIGIN
            ).set_color(color)

        # Create arrows only for actual cycles
        all_arrows = VGroup()
        for cycle_info in perm_cycles:
            indices, color, angle = cycle_info  # Proper unpacking
            if len(indices) == 1: continue  # Skip fixed points
            
            cycle_arrows = VGroup()
            for i in range(len(indices)):
                start = indices[i]
                end = indices[(i+1)%len(indices)]
                cycle_arrows.add(create_arrow(start, end, color, angle))
            all_arrows.add(cycle_arrows)

        # Animation sequence
        self.play(FadeIn(dots, lag_ratio=0.1, run_time=2))
        self.wait(0.5)

        # Animate cycles
        self.play(LaggedStart(
            *[Create(cycle) for cycle in all_arrows],
            lag_ratio=0.3,
            run_time=4
        ))

        # Handle fixed point separately
        fixed_dot = dots[8]
        self.play(
            fixed_dot[0].animate.set_color(BLUE),
            run_time=1
        )
        self.wait(2)
        self.play(FadeOut(all_arrows, dots))
```
