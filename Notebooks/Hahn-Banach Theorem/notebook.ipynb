```python
%%manim -qm S5

class S5(Scene):
    def construct(self):
        svg = SVGMobject("linear.svg", height=8)
        subs = svg.submobjects
        #numerateSVG("linear.svg", self)

        halfL = subs[0]
        halfL.z_index=-1
        halfR = subs[1]
        halfR.z_index=-1
        convex = subs[2]
        point = subs[3]
        line = subs[4]
        equal = subs[5]
        less = subs[6]
        greater = subs[7]

        self.play(FadeIn(convex, point))
        self.play(Create(line))
        self.wait()
        equ = MathTex("Ax + By = C", font_size=60).move_to(DOWN*3 + RIGHT*3.5)
        self.play(FadeIn(equ))
        self.wait()

        defs = MathTex(r"F:\mathbb{R}^2 \to \mathbb{R}, F(x,y) = Ax + By", font_size=60).to_edge(UP)
        self.play(FadeOut(convex, point, line), ReplacementTransform(equ, defs))
        self.wait()

        lin1 = Tex(r"$F$ is linear, so $F(a\vec v_1 + b\vec v_2) = aF(\vec v_1) + bF(\vec v_2)$.", font_size=60).next_to(defs, DOWN).shift(DOWN*0.5)
        lin2 = Tex(r"$\{\vec v \in \mathbb{R}^2 : F(\vec v) = C\}$ is a line.", font_size=60).next_to(lin1, DOWN).shift(DOWN*0.5)
        self.play(Write(lin1))
        self.wait()
        self.play(FadeIn(lin2))
        self.wait()
        self.play(FadeOut(lin1, lin2, defs))

        imgs = [ImageMobject(f"generalset{i}.png") for i in range(1,5)]
        self.play(FadeIn(imgs[0], imgs[3]))
        self.wait()
        hyp = Tex(r"A ", r"hyperplane", r" in $X$ is a set of the form $\{x \in X : F(x) = C\}$ \\ for some ", r"linear", r" $F:X \to \mathbb{R}$ with $F\ne 0$ and $C \in \mathbb{R}$.")
        hyp[1].set_color("#ccff00ff")
        hyp[3].set_color("#e5ff80ff")
        hyp.shift(UP*2.3)
        lineardef = VGroup(
            Tex(r"$F(x + y) = F(x) + F(y)$ for all $x,y \in X$"),
            Tex(r"$F(ax) = aF(x)$ for all $x\in X$ and $a\in \mathbb{R}$")
        ).arrange(DOWN).next_to(hyp, DOWN)
        self.play(Write(hyp))
        
        self.wait()
        self.play(FadeIn(lineardef))
        self.wait()
        
        self.play(FadeOut(imgs[0], imgs[3], lineardef, hyp))

        self.play(FadeIn(convex, point, line))
        self.wait()
        self.play(FadeIn(equal))
        self.wait()
        self.play(FadeIn(halfL, halfR))
        self.wait()
        self.play(FadeIn(less))
        self.play(FadeIn(greater))
        self.wait()

        seperatedef = Tex(r"A ", r"linear", r" functional $F: X \to \mathbb{R}$ is said to separate $K$ and $x_0$ \\ if $F(x) \le C$ for $x \in K$ and $F(x_0) \ge C$ for a $C \in \mathbb{R}$.")
        seperatedef.to_edge(UP)
        rect = SurroundingRectangle(seperatedef, color = WHITE, fill_color=BLACK, fill_opacity=1, buff=0.15)
        seperatedef[1].set_color("#e5ff80ff")
        self.play(FadeIn(rect))
        self.play(Write(seperatedef))
        self.wait()
        self.play(FadeOut(seperatedef, rect, *subs))

        
```

```python
%%manim -qm S6

class S6(Scene):
    def construct(self):
        img = ImageMobject("sublinear1.png")
        

        defs = VGroup(
            Tex(r"$p:X \to \mathbb{R}$ is ", r"sublinear", r" if", font_size=60),
            Tex(r"$p(x + y) \le p(x) + p(y)$ for all $x,y \in X$", font_size=60),
            Tex(r"$p(ax) = ap(x)$ for all $x \in X$ and $a \ge 0$", font_size=60)
        ).arrange(DOWN).to_edge(UP).shift(DOWN*0.5)
        defs[0][1].set_color("#ffe680ff")

        line = Line(start = LEFT*4, end = RIGHT*4, color=WHITE)

        defs2 = VGroup(
            Tex(r"$F:X \to \mathbb{R}$ is ", r"linear", r" if", font_size=60),
            Tex(r"$F(x + y) = F(x) + F(y)$ for all $x,y \in X$", font_size=60),
            Tex(r"$F(ax) = aF(x)$ for all $x\in X$ and $a\in \mathbb{R}$", font_size=60)
        ).arrange(DOWN).to_edge(DOWN).shift(UP*0.5)
        defs2[0][1].set_color("#e5ff80ff")
        
        self.play(FadeIn(defs))
        self.wait()
        self.play(FadeIn(line, defs2))
        self.wait()
        self.play(FadeOut(defs, line, defs2))
        
        self.play(FadeIn(img))
        self.wait()
        ball = VGroup(Dot(), Tex(r"0"), Circle(radius=0.7, color=WHITE))
        ball[1].next_to(ball[0], buff=0.1)
        ball.shift(DOWN*1.45 + LEFT*0.75)
        eps = Tex(r"$B_\varepsilon(0) \subseteq K$", font_size=70).to_edge(UP).shift(RIGHT*5)
        self.play(FadeIn(ball, eps))
        self.wait()
        
        self.play(FadeOut(img, ball, eps))

        min1 = Tex(r"Let $K \subseteq X$ be convex, such that $B_\varepsilon(0) \subseteq K$ for some $\varepsilon > 0$.")
        min2= Tex(r"Define the Minkowski functional $p_K:X \to \mathbb{R}$ \\ by $p_K(x) = \inf\{r > 0 : x \in rK\}$")
        min3 = Tex(r"Then $p_K$ is sublinear and \\ satisfies $p_K(x) \le 1$ if $x \in K$ and $p_K(x) \ge 1$ if $x \notin K$.")
        mins = VGroup(min1, min2, min3).arrange(DOWN, buff=0.3)
        self.play(FadeIn(min1))
        self.wait()
        self.play(Write(min2))
        self.wait()
        self.play(Write(min3))
        self.wait()
        self.play(FadeOut(mins))
```
