```python
%%manim -qh Thumbnail

class Thumbnail(Scene):
    def construct(self):
        title = Tex("Title", font_size=165)
        #title.set_color_by_gradient(*gradient_colors)
        self.add(title)
```

```python
def numerateSVG(path, scene, start=0):
    svg = SVGMobject("seperating1.svg", height=8)
    subs = svg.submobjects
    for i in range(start, len(subs)):
        scene.add(subs[i])
        scene.add(Text(f"{i}").next_to(subs[i]))
        scene.wait()
```

```python
%%manim -qm S1

class S1(Scene):
    def construct(self):
        svg = SVGMobject("seperating1.svg", height=8)
        subs = svg.submobjects
        #for i in range(len(subs)):
        #    self.add(subs[i])
        #    self.add(Text(f"{i}").next_to(subs[i]))
        #    self.wait()

        self.play(Create(subs[0]), FadeIn(subs[7], subs[8]))
        self.wait()
        self.play(FadeIn(subs[1], subs[2]))
        self.play(Create(subs[3]))
        self.wait()
        self.play(FadeIn(subs[4], subs[6]))

        self.wait()
        self.play(Create(subs[5]), FadeIn(subs[9]))
        self.wait()
        self.play(FadeOut(*subs[:-1]))
```

```python
%%manim -qm S2

class S2(Scene):
    def construct(self):
        svg = SVGMobject("seperating2.svg", height=8)
        subs = svg.submobjects
        #for i in range(len(subs)):
        #    self.add(subs[i])
        #    self.add(Text(f"{i}").next_to(subs[i]))
        #    self.wait()


        self.play(Create(subs[3]), FadeIn(subs[5]))
        
        self.play(FadeIn(subs[6], subs[7]))
        self.play(Create(subs[8]))
        self.wait()
        
        self.play(FadeIn(*subs[0:3]))
        self.wait()

        self.play(Create(subs[4]))
        self.wait()
        self.play(FadeOut(*subs[:-1]))
```

```python
%%manim -qm S3

class S3(Scene):
    def construct(self):
        numerateSVG("seperating2.svg", self)
        
```
