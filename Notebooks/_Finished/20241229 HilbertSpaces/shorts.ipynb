```python
from manim import *
config.media_width = "40%"
config.verbosity = "WARNING"

config.frame_size = (1080, 1920) 
config.frame_width = 6

gradient_colors = ["#FAD961", "#F76B1C"]
```

```python
%%manim Intro

class Intro(Scene):
    def construct(self):
        tex1 = Tex(r"Hilbert spaces", font_size=85).set_color_by_gradient(*gradient_colors).shift(UP*1.5)
        tex2 = Tex(r"explained", font_size=80).next_to(tex1, DOWN, buff=0.25)
        tex3 = Tex(r"in $1$ minute", font_size=85).next_to(tex2, DOWN, buff=0.25)

        self.add(tex1, tex2, tex3)
        self.wait()
```

```python
%%manim DistancesAndAnglesScrambled

class DistancesAndAnglesScrambled(Scene):
    def construct(self):
        eucl = Tex(r"Euclidean Spaces", font_size=75).set_color_by_gradient(*gradient_colors).shift(UP*0.3)
        self.play(FadeIn(eucl))
        
        # ----- Top: Distance -----
        # Scrambled points for distance
        P1 = Dot(LEFT * 2 + UP * 4)
        P2 = Dot(RIGHT * 1.5 + UP * 3.5)

        # Double Arrow for distance
        distance_arrow = DoubleArrow(P1.get_center(), P2.get_center(), buff=0.1, color=BLUE)
        distance_label = MathTex("d").next_to(distance_arrow, UP)

        # Text "Distances" below
        distances_text = Tex("Distances", font_size=80).next_to(distance_arrow, DOWN, buff=1)

        # ----- Bottom: Angle -----
        # Scrambled points for angle
        A = Dot(LEFT * 1.5 + DOWN * 1)
        B = Dot(DOWN * 3.5 + UP*1)
        C = Dot(RIGHT * 2 + DOWN * 1.5)

        # Lines forming the angle
        line_AB = Line(B.get_center(), A.get_center())
        line_BC = Line(B.get_center(), C.get_center())

        # Angle at B
        angle = Angle(line_BC, line_AB, radius=0.5)
        angle_label = MathTex(r"\theta").move_to(B.get_center() + 0.8 * UP)

        # Text "Angles" below
        angles_text = Tex("Angles", font_size=80).next_to(B, DOWN, buff=1)

        # ----- Animation -----
        self.add(P1, P2)
        self.play(GrowArrow(distance_arrow), FadeIn(distance_label), FadeIn(distances_text))

        self.add(A, B, C)
        self.play(Create(line_AB), Create(line_BC))
        self.play(Create(angle), FadeIn(angle_label), FadeIn(angles_text))

        self.wait()
```

```python
%%manim S1

class S1(Scene):
    def construct(self):
        tex = Tex("Orthonormal basis", font_size=65).move_to(UP*3.5)
        basis_vectors2 = VGroup(
            Vector([1,0],color=RED),
            Vector([0,1],color=GREEN)
        )
        labels2 = VGroup(
            MathTex(r"\vec{b}_1", color=RED).next_to(basis_vectors2[0].get_end(), DOWN),
            MathTex(r"\vec{b}_2", color=GREEN).next_to(basis_vectors2[1].get_end(), LEFT)
        )
        plane2 = NumberPlane(y_range=(-20,2,1), x_range=(-10,2, 1))
        plane2.set_z_index(-1)
        self.play(Create(basis_vectors2), Create(labels2), Write(tex))
        self.wait()
        self.play(Create(plane2))
        self.wait()

        vec1 = Vector([2, -3])
        lab1 = MathTex(r"\begin{pmatrix}2\\ -3\end{pmatrix}").next_to(vec1.get_end() + LEFT*0.2, LEFT)
        tex2 = Tex(r"Unique Coordinates", font_size=60).move_to(DOWN*4.5)
        self.play(Create(vec1), Create(lab1))
        self.play(FadeIn(tex2))
        self.wait()

        norm1 = MathTex(r"\|v\| = \sqrt{\sum_{i=1}^n |v_i|^2}", font_size=70).move_to(UP*3.5)
        self.play(FadeOut(tex), FadeIn(norm1))
        self.wait()

        self.play(FadeOut(basis_vectors2, plane2, vec1, lab1, tex2, labels2))
        fin1 = Tex(r"finite-dimensional", font_size=60).shift(UP*1.5)
        arr = Arrow(start = UP*1.5, end = DOWN*1)
        labarr = MathTex(r"n \to \infty").next_to(arr, RIGHT)
        fin2 = Tex(r"infinite-dimensional", font_size=60).shift(DOWN*1).set_color_by_gradient(*gradient_colors)

        self.play(FadeIn(fin1))
        self.play(GrowArrow(arr), FadeIn(labarr))
        self.play(FadeIn(fin2))
        self.wait()

        norm2 = MathTex(r"\|v\| = \sqrt{\sum_{i=1}^\infty |v_i|^2}", font_size=70).move_to(UP*3.5)
        self.play(Transform(norm1, norm2))
        self.wait()

        inf = Tex(r"$\sqrt{\sum_{i=1}^\infty |v_i|^2} < \infty$, but \\no vector with \\coordinates $(v_1, v_2, v_2, \ldots)$").next_to(fin2, DOWN, buff=0.5)
        self.play(FadeIn(inf))
        self.wait()

        hilbert = Tex(r"Hilbert space", font_size=75).set_color_by_gradient(*gradient_colors).move_to(UP*0.75)
        self.play(ReplacementTransform(VGroup(fin1, labarr, fin2, arr), hilbert))
        self.wait()
        hildef = Tex(r"complete\\ inner product space").next_to(hilbert, DOWN)
        inner = MathTex(r"\langle v, w \rangle = \sum_{i=1}^\infty v_i \overline{w_i}", font_size=70).next_to(hildef, DOWN)
        self.play(FadeOut(inf), FadeIn(hildef))
        self.play(FadeIn(inner))
        self.wait()

        
```
