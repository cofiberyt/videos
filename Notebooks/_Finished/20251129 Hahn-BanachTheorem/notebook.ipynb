```python
from manim import *
import numpy as np
config.media_width = "75%"
config.verbosity = "WARNING"

def numerateSVG(path, scene, start=0):
    svg = SVGMobject(path, height=8)
    subs = svg.submobjects
    for i in range(start, len(subs)):
        scene.add(subs[i])
        scene.add(Text(f"{i}").next_to(subs[i]))
        scene.wait()

# colors
```

```python
%%manim -qh S5

class S5(Scene):
    def construct(self):
        svg = SVGMobject("linear.svg", height=8)
        subs = svg.submobjects
        #numerateSVG("linear.svg", self)

        halfL = subs[0]
        halfL.z_index=-1
        halfR = subs[1]
        halfR.z_index=-1
        convex = subs[2]
        point = subs[3]
        line = subs[4]
        equal = subs[5]
        less = subs[6]
        greater = subs[7]

        self.play(FadeIn(convex, point))
        self.play(Create(line))
        self.wait()
        equ = MathTex("Ax + By = C", font_size=60).move_to(DOWN*3 + RIGHT*3.5)
        self.play(FadeIn(equ))
        self.wait()

        defs = MathTex(r"F:\mathbb{R}^2 \to \mathbb{R}, F(x,y) = Ax + By", font_size=60).to_edge(UP)
        self.play(FadeOut(convex, point, line), ReplacementTransform(equ, defs))
        self.wait()

        lin1 = Tex(r"$F$ is linear, so $F(\vec v_1 + \vec v_2) = F(\vec v_1) + F(\vec v_2)$ and\\ $F(a\vec v) = aF(\vec v)$.", font_size=60).next_to(defs, DOWN).shift(DOWN*0.5)
        lin2 = Tex(r"$\{\vec v \in \mathbb{R}^2 : F(\vec v) = C\}$ is a line.", font_size=60).next_to(lin1, DOWN).shift(DOWN*0.5)
        self.play(Write(lin1))
        self.wait()
        self.play(FadeIn(lin2))
        self.wait()
        self.play(FadeOut(lin1, lin2, defs))

        imgs = [ImageMobject(f"generalset{i}.png") for i in range(1,5)]
        self.play(FadeIn(imgs[0], imgs[3]))
        self.wait()
        hyp = Tex(r"A ", r"hyperplane", r" in $X$ is a set of the form $\{x \in X : F(x) = C\}$ \\ for some ", r"linear", r" $F:X \to \mathbb{R}$ with $F\ne 0$ and $C \in \mathbb{R}$.")
        hyp[1].set_color("#ccff00ff")
        hyp[3].set_color("#e5ff80ff")
        hyp.shift(UP*2.3)
        lineardef = VGroup(
            Tex(r"$F(x + y) = F(x) + F(y)$ for all $x,y \in X$"),
            Tex(r"$F(ax) = aF(x)$ for all $x\in X$ and $a\in \mathbb{R}$")
        ).arrange(DOWN).next_to(hyp, DOWN)
        self.play(Write(hyp))
        
        self.wait()
        self.play(FadeIn(lineardef))
        self.wait()
        
        self.play(FadeOut(imgs[0], imgs[3], lineardef, hyp))

        self.play(FadeIn(convex, point, line))
        self.wait()
        self.play(FadeIn(equal))
        self.wait()
        self.play(FadeIn(halfL, halfR))
        self.wait()
        self.play(FadeIn(less))
        self.play(FadeIn(greater))
        self.wait()

        seperatedef = Tex(r"A ", r"linear", r" functional $F: X \to \mathbb{R}$ is said to separate $K$ and $x_0$ \\ if $F(x) \le C$ for $x \in K$ and $F(x_0) \ge C$ for a $C \in \mathbb{R}$.")
        seperatedef.to_edge(UP)
        rect = SurroundingRectangle(seperatedef, color = WHITE, fill_color=BLACK, fill_opacity=1, buff=0.15)
        seperatedef[1].set_color("#e5ff80ff")
        self.play(FadeIn(rect))
        self.play(Write(seperatedef))
        self.wait()
        self.play(FadeOut(seperatedef, rect, *subs))

        
```

```python
%%manim -qh S6

class S6(Scene):
    def construct(self):
        img = ImageMobject("sublinear1.png")
        

        defs = VGroup(
            Tex(r"$p:X \to \mathbb{R}$ is ", r"sublinear", r" if", font_size=60),
            Tex(r"$p(x + y) \le p(x) + p(y)$ for all $x,y \in X$", font_size=60),
            Tex(r"$p(ax) = ap(x)$ for all $x \in X$ and $a \ge 0$", font_size=60)
        ).arrange(DOWN).to_edge(UP).shift(DOWN*0.5)
        defs[0][1].set_color("#ffe680ff")

        line = Line(start = LEFT*4, end = RIGHT*4, color=WHITE)

        defs2 = VGroup(
            Tex(r"$F:X \to \mathbb{R}$ is ", r"linear", r" if", font_size=60),
            Tex(r"$F(x + y) = F(x) + F(y)$ for all $x,y \in X$", font_size=60),
            Tex(r"$F(ax) = aF(x)$ for all $x\in X$ and $a\in \mathbb{R}$", font_size=60)
        ).arrange(DOWN).to_edge(DOWN).shift(UP*0.5)
        defs2[0][1].set_color("#e5ff80ff")
        
        self.play(FadeIn(defs))
        self.wait()
        self.play(FadeIn(line, defs2))
        self.wait()
        self.play(FadeOut(defs, line, defs2))
        
        self.play(FadeIn(img))
        self.wait()
        ball = VGroup(Dot(), Tex(r"0"), Circle(radius=0.7, color=WHITE))
        ball[1].next_to(ball[0], buff=0.1)
        ball.shift(DOWN*1.45 + LEFT*0.75)
        eps = Tex(r"$B_\varepsilon(0) \subseteq K$", font_size=70).to_edge(UP).shift(RIGHT*5)
        self.play(FadeIn(ball, eps))
        self.wait()
        
        self.play(FadeOut(img, ball, eps))

        min0 = Tex(r"Minkowski functional", font_size=75).move_to(UP*2.75)
        min1 = Tex(r"Let $K \subseteq X$ be convex, such that $B_\varepsilon(0) \subseteq K$ for some $\varepsilon > 0$.")
        min2= Tex(r"Define the Minkowski functional $p_K:X \to \mathbb{R}$ \\ by $p_K(x) = \inf\{r > 0 : x \in rK\}$")
        min3 = Tex(r"Then $p_K$ is sublinear and \\ satisfies $p_K(x) \le 1$ if $x \in K$ and $p_K(x) \ge 1$ if $x \notin K$.")
        mins = VGroup(min1, min2, min3).arrange(DOWN, buff=0.3)
        self.play(FadeIn(min0, mins))
        self.wait()
        self.play(FadeOut(min0, mins))
```

```python
%%manim -qh S7

class S7(Scene):
    def construct(self):
        im1 = ImageMobject("linear.png").scale(0.5)
        im2 = ImageMobject("sublinear1.png").scale(0.5)
        ims = Group(im1, im2).arrange(LEFT, buff=0).to_edge(UP, buff=0)
        
        self.play(FadeIn(ims))
        self.wait()

        tex0 = Tex(r"$F$ is dominated by $p_K$, if $F \le p_K$, \\", r"so $F(x) \le p_K(x)$ for all $x \in X$")
        tex0.shift(DOWN*0.8)
        self.play(Write(tex0))
        self.wait()

        tex3 = Tex(r"$F(x) \le p_K(x) \le C$ if $x \in K$", r" and $F(x_0) = p_K(x_0) \ge C$").shift(DOWN*2.5)
        self.play(FadeIn(tex3[0]))
        self.wait()
        self.play(FadeIn(tex3[1]))
        self.wait()
        self.play(FadeOut(tex3, tex0))
        
        tex1 = Tex(r"Suppose $F:X \to \mathbb{R}$ is a linear functional such that \\", r"$F \le p_K$ and $F(x_0) = p_K(x_0)$, then $F$ separates $K$ and $x_0$.")
        tex1.shift(DOWN*0.8)
        self.play(Write(tex1))
        self.wait()
        
        tex4 = Tex(r"Let $Y = \text{span}(x_0)$. Then $F(ax_0) = ap_K(x_0)$. \\", r" $F$ has to extend the linear functional $f_0:Y\to \mathbb{R}$ \\ defined by $f_0(ax_0) = ap_K(x_0).$ ($f_0 \le p_K$)")
        tex4.to_edge(DOWN)
        self.play(FadeIn(tex4[0]))
        self.wait()
        self.play(Write(tex4[1]))
        self.wait()
        self.play(FadeOut(ims, tex1, tex4))
```

```python
%%manim -qh S8

class S8(Scene):
    def construct(self):
        title = Tex(r"Hahn-Banach theorem", font_size=70)
        tex1 = Tex(r"Let $X$ be a real vector space, $Y \subseteq X$ a subspace,\\ $f_0: Y \to \mathbb{R}$ a ", r"linear", r" functional and  $p:X \to \mathbb{R}$", r" sublinear", r", \\ such that $f_0 \le p$.", font_size=53)
        tex2 = Tex(r"Then there exists a ", r"linear extension", r" $F:X \to \mathbb{R}$ of $f_0$, \\ such that $F \le p$.", font_size=53)
        title.set_color("#ff555555")
        tex1[1].set_color("#ccff00ff")
        tex1[3].set_color("#ffcc00ff")
        tex2[1].set_color("#ff555555")
        
        title.shift(UP*2.5)
        tex1.shift(UP*0.9) 
        tex2.shift(DOWN*1)
        
        self.play(FadeIn(title))
        self.wait()
        self.play(FadeIn(tex1))
        self.wait()
        self.play(Write(tex2))
        self.wait()
        self.play(FadeOut(title, tex1, tex2))
```

```python
%%manim -qh S9

class S9(Scene):
    def construct(self):
        # Hahn-Banach for normed spaces
        # point --> line, line --> plane
        title = Tex(r"Hahn-Banach continuous extension theorem", font_size=65)
        tex1 = Tex(r"Let $X$ be a ", r"normed space", r", $Y \subset X$ a subspace \\ and $f_0:Y\to \mathbb{R}$ a ", r"continuous linear", r" functional.", font_size=55)
        tex2 = Tex(r"Then there exists a ", r"continuous linear", r" extension \\ $F:X \to \mathbb{R}$ of $f_0$ that satisfies $\|f_0\| = \|F\|$.", font_size=55)
        title.set_color("#ff555555")
        tex1[1].set_color("#00ccff55")
        tex1[3].set_color("#ccff00ff")
        tex2[1].set_color("#ccff00ff")

        title.shift(UP*2.5)
        tex1.shift(UP*1) 
        tex2.shift(DOWN*0.75)

        self.play(FadeIn(title))
        self.wait()
        self.play(FadeIn(tex1))
        self.wait()
        self.play(Write(tex2))
        self.wait()

        
        dist = Tex(r"$1/\|f_0\|$ is the distance of the hyperplane \\ $\{x \in Y : f_0(x) = 1\}$ from the origin.").shift(DOWN*2.75)
        ext = Tex(r"Extend hyperplane $\{x \in Y : f_0(x) = 1\}$ in $Y$ to \\ a hyperplane $\{x \in X : F(x) = 1\}$ in $X$, \\which has the same distance to the origin.")
        ext.shift(DOWN*2.8)
        
        self.play(FadeIn(dist))
        self.wait()
        self.play(Transform(dist, ext))
        self.wait()
        self.play(FadeOut(tex1, tex2, dist))

        
        
        arr1 = Arrow(start=LEFT, end=RIGHT*4)
        arr2 = Arrow(start=DOWN, end=UP*2.5)
        point = Dot(radius=0.1).set_color(RED).move_to(RIGHT*2)
        line = Line(start=RIGHT*2+DOWN, end=RIGHT*2+UP*2).set_color(RED)
        coor1 = VGroup(arr1, point, line, arr2).move_to(DOWN*2).shift(LEFT*4)
        coor2 = coor1.copy().shift(RIGHT*8)

        lab1 = Tex(r"Continuous linear \\ functional $f_0$ on $\mathbb{R}$").move_to(LEFT*4 + UP*0.75)
        lab2 = Tex(r"Continuous linear \\ extension $F$ on $\mathbb{R}^2$ of $f_0$").move_to(RIGHT*4 + UP*0.75)
        extend = Arrow(start=LEFT*1.75 + UP*0.75, end=RIGHT*1.25+UP*0.75)
        

        self.play(FadeIn(coor1[0:2], lab1))
        self.wait()
        self.play(GrowArrow(extend))
        self.wait()
        self.play(FadeIn(coor2, lab2))
        self.wait()        
        self.play(FadeOut(coor1[0:2], coor2, lab1, lab2, extend, title))
```
