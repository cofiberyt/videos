```python
%%manim -qh S1

class S1(Scene):
    def construct(self):
        series = MathTex(r"\sum_{k=1}^\infty \frac{1}{k^{2n}} = 1 + \frac{1}{4^n} + \frac{1}{9^n} + \frac{1}{16^n} + \frac{1}{25^n} + \cdots")
        euler = Tex(r"Euler")
        ex1 = MathTex(r"\sum_{k=1}^\infty \frac{1}{k^2} = \frac{\pi^2}{6}")
        ex2 = MathTex(r"\sum_{k=1}^\infty \frac{1}{k^4} = \frac{\pi^4}{90}")
        ex3 = MathTex(r"\sum_{k=1}^\infty \frac{1}{k^6} = \frac{\pi^6}{945}")
        ex = VGroup(ex1, ex2, ex3)
        ex1.next_to(ex2, UP)
        ex3.next_to(ex2, DOWN)
        
        euler.move_to(DOWN*3.5)
        ex.move_to(RIGHT*3.5 + DOWN*0.7)
        
        self.play(FadeIn(series))
        self.wait()
        self.play(series.animate.shift(UP*3))
        self.play(FadeIn(euler))
        self.wait()
        self.play(euler.animate.shift(LEFT*3))
        self.play(Write(ex))
        self.wait()
        self.play(FadeOut(euler, ex))

        series2 = MathTex(r"\sum_{k=1}^\infty \frac{1}{(2k-1)^{2n}} = 1 + \frac{1}{9^n} + \frac{1}{25^n} + \frac{1}{49^n} + \cdots")
        series3 = MathTex(r"\sum_{k=1}^\infty (-1)^{k+1}\frac{1}{k^{2n}} = 1 - \frac{1}{4^n} + \frac{1}{9^n} - \frac{1}{16^n} + \frac{1}{25^n} - \cdots")
        series4 = MathTex(r"\sum_{k=1}^\infty\frac{(-1)^{k+1}}{(2k-1)^{2n+1}} = 1 - \frac{1}{3^{2n+1}} + \frac{1}{5^{2n+1}} - \frac{1}{7^{2n+1}} + \cdots")
        series2.move_to(UP*1)
        series3.move_to(DOWN*1)
        series4.move_to(DOWN*3)
        self.play(FadeIn(series2, series3, series4))
        self.wait()
        self.play(FadeOut(series2, series3, series4))
        self.wait()

        squ = MathTex(r"\sum_{k=1}^\infty \frac{1}{k^2}", r" = 1 + \frac{1}{2\cdot 2} + \frac{1}{3 \cdot 3} + \frac{1}{4 \cdot 4} + \frac{1}{5\cdot 5} + \cdots")
        squ.move_to(UP*3)
        le = MathTex(r"< 1 + \frac{1}{1\cdot 2} + \frac{1}{2 \cdot 3} + \frac{1}{3 \cdot 4} + \frac{1}{4 \cdot 5} + \cdots")
        le2 = MathTex(r"= 1 + \frac{2-1}{1\cdot 2} + \frac{3-2}{2 \cdot 3} + \frac{4-3}{3 \cdot 4} + \frac{5-4}{4 \cdot 5} + \cdots")
        le3 = MathTex(r"= 1 + 1 - \frac{1}{2} + \frac{1}{2} - \frac{1}{3} + \frac{1}{3} - \frac{1}{4} + \frac{1}{4} - \frac{1}{5} + \cdots")
        le4 = MathTex(r"= 2")
        les = VGroup(le, le2, le3, le4).arrange(DOWN, aligned_edge=LEFT)
        les.next_to(squ, DOWN).shift(RIGHT*1.5)
        
        self.play(ReplacementTransform(series, squ))
        self.wait()
        self.play(FadeIn(le))
        self.wait()
        self.play(FadeIn(le2))
        self.wait()
        self.play(FadeIn(le3))
        self.wait()
        self.play(FadeIn(le4))
        self.wait()
        
        final1 = MathTex(r"\sum_{k=1}^\infty \frac{1}{k^2} < 2")
        self.play(ReplacementTransform(VGroup(les, squ), final1))
        self.wait()
        final2 = MathTex(r"\sum_{k=1}^\infty \frac{1}{k^{2n}} < 2")
        self.play(ReplacementTransform(final1, final2))
        self.wait()
        self.play(FadeOut(final2))
```

```python
%%manim -qh S2

class S2(Scene):
    def construct(self):
        gen1 = Tex(r"Generating function")
        gen2 = MathTex(r"f(z) = ", r"\sum_{n=1}^\infty", r" \left(", r"\sum_{k=1}^\infty", r" \frac{1}{k^{2n}}", r" \right)", r"z^{2n}")
        gen1.move_to(UP*2.5)
        gen2.next_to(gen1, DOWN)

        num = Tex(r"Number Theory")
        ana = Tex(r"Complex Analysis")
        arr = Arrow(1.5*LEFT, 1.5*RIGHT, color=BLACK)
        arr.move_to(DOWN*1)
        num.next_to(arr, LEFT)
        ana.next_to(arr, RIGHT)

        self.play(Write(gen1))
        self.play(FadeIn(gen2))
        self.wait()
        self.play(Write(num))
        self.play(GrowArrow(arr))
        self.play(Write(ana))
        self.wait()

        self.play(FadeOut(num, arr, ana, gen1), gen2.animate.move_to(ORIGIN))
        self.wait()
        
        possum1 = gen2[1].get_center()
        possum2 = gen2[3].get_center()

        self.play(gen2[1].animate.move_to(possum2), gen2[3].animate.move_to(possum1))
        self.wait()

        new = MathTex(r"\frac{z^{2n}}{k^{2n}}")
        pos4 = gen2[4].get_center()
        new.move_to(pos4)
        self.play(gen2[6].animate.move_to(pos4 + UP*0.35), Transform(gen2[4], new))
        self.remove(gen2[6])
        self.wait()
        self.play(gen2[0:6].animate.shift(UP*2.5))

        geo = MathTex(r"\sum_{n=0}^\infty q^n = \frac{1}{1-q} \text{ if } |q|<1")
        geo.move_to(DOWN*0.5)
        self.play(Write(geo))
        self.wait()
        inner = VGroup(gen2[1], gen2[4]).copy()
        self.play(geo.animate.shift(DOWN*2.5))
        self.play(inner.animate.move_to(ORIGIN))
        self.wait()

        inner2 = MathTex(r"= \frac{\frac{z^2}{k^2}}{1 - \frac{z^2}{k^2}}")
        inner3 = MathTex(r"= \frac{z^2}{k^2 - z^2}")
        
        self.play(inner.animate.shift(LEFT*1.5))
        inner2.next_to(inner, RIGHT)
        inner3.next_to(inner, RIGHT)
        self.play(FadeIn(inner2))
        self.wait()
        self.play(Transform(inner2, inner3))
        self.wait()

        gen3 = MathTex(r"f(z) ", r"= \sum_{k=1}^\infty \frac{z^2}{k^2 - z^2}")
        self.play(FadeOut(gen2[0:6], inner, inner2, geo))
        self.play(FadeIn(gen3))
        self.wait()
        
        gen4 = MathTex(r"= -\frac{z}{2}\sum_{k=1}^\infty \frac{2z}{z^2-k^2}")
        self.play(gen3.animate.shift(LEFT*2.5))
        gen4.next_to(gen3, RIGHT)
        self.play(Write(gen4))
        self.wait()

        gen5 = MathTex(r"= -\frac{z}{2}\sum_{k=1}^\infty \left( \frac{1}{z-k} + \frac{1}{z+k}\right)")
        self.play(gen4.animate.next_to(gen3[0], RIGHT), FadeOut(gen3[1]))
        gen5.next_to(gen4, RIGHT)
        self.play(Write(gen5))
        self.wait()

        self.play(gen5.animate.next_to(gen3[0], RIGHT), FadeOut(gen4))
        self.wait()
        cgen = VGroup(gen3[0], gen5)
        self.play(cgen.animate.move_to(UP*3))
        self.wait()

        poles1 = Tex(r"Pols at every integer $\ne 0$")
        poles1.next_to(cgen, DOWN)
        corr = MathTex(r"\frac{1}{z} - 2\frac{f(z)}{z} = \lim_{n\to\infty} \sum_{k=-n}^n \frac{1}{z-k} = \frac{1}{z} + \sum_{k=1}^\infty \frac{2z}{z^2-k^2}")
        corr.move_to(DOWN)
        poles2 = Tex(r"Poles at every integer of residue $1$")
        poles2.next_to(corr, DOWN)

        self.play(Write(poles1))
        self.wait()
        self.play(FadeIn(corr))
        self.play(FadeIn(poles2))
        self.wait()
        corpol = VGroup(corr, poles2)

        self.play(FadeOut(cgen, poles1), corpol.animate.shift(UP*3.7))
        self.wait()

        perio1 = MathTex(r"\cdots + \frac{1}{z+3} + \frac{1}{z+2} + \frac{1}{z+1} + \frac{1}{z} + \frac{1}{z-1} + \frac{1}{z-2} + \frac{1}{z-3} + \cdots")
        #perio1.move_to(UP*0.5)
        self.play(FadeIn(perio1))
        arrpio = Arrow(DOWN*0.5, DOWN*2, color=BLACK)
        periotr = MathTex(r"z \leadsto z+1")
        periotr.next_to(arrpio, RIGHT)
        perio2 = MathTex(r"\cdots + \frac{1}{z+4} + \frac{1}{z+3} + \frac{1}{z+2} + \frac{1}{z+1} + \frac{1}{z} + \frac{1}{z-1} + \frac{1}{z-2} + \cdots")
        perio2.move_to(DOWN*2.5)
        self.play(GrowArrow(arrpio), FadeIn(periotr))
        self.play(FadeIn(perio2))
        self.wait()
        perio3 = Tex(r"periodic with period $1$")
        perio3.next_to(perio2, DOWN)
        self.play(Write(perio3))
        self.wait()
        self.play(FadeOut(perio1, arrpio, periotr, perio2), perio3.animate.next_to(poles2, DOWN))
        self.wait()

        cot = MathTex(r"\pi \cot \pi z = ", r"\frac{\pi \cos \pi z}{\sin \pi z}")
        cot.move_to(DOWN*1)
        self.play(FadeIn(cot))
        self.wait()

        cot2 = MathTex(r"\frac{\frac{d}{dz} \sin \pi z}{\sin \pi z}")
        cot2.next_to(cot[0], RIGHT)
        self.play(Transform(cot[1], cot2))
        self.wait()

        loga = Tex(r"$\leadsto$ Logarithmic derivative")
        loga.next_to(cot, DOWN)
        self.play(Write(loga))
        self.wait()
        self.play(FadeOut(corr, poles2, perio3, cot, loga))
```

```python
%%manim -qh S3

class S3(Scene):
    def construct(self):
        title = Tex(r"Logarithmic derivate")
        title.move_to(UP*3)
        pos = MathTex(r"f > 0 \Rightarrow \frac{d}{dz} \ln(f(z)) = \frac{f'(z)}{f(z)}")
        pos.next_to(title, DOWN)
        def1 = MathTex(r"L(f) := \frac{f'}{f}")
        def1.next_to(title, DOWN)

        self.play(FadeIn(title))
        self.wait()
        self.play(FadeIn(pos))
        self.wait()
        self.play(Transform(pos, def1))
        self.wait()

        mult = MathTex(r"L(f \cdot g) = \frac{f'g + fg'}{fg} = L(f) + L(g)")
        mult.move_to(UP*0.5)
        mult2 = Tex(r"Under certain conditions")
        mult3 = MathTex(r"L\left( \prod_{k=1}^\infty f_k \right) = \sum_{k=1}^\infty L(f_k)")
        mult2.move_to(DOWN*1.5)
        mult3.next_to(mult2, DOWN)
        inf = VGroup(mult2, mult3)
        tit = VGroup(title, pos)
        
        self.play(Write(mult))
        self.wait()
        self.play(FadeIn(mult2, mult3))
        self.wait()
        self.play(FadeOut(mult), tit.animate.move_to(UP*2.7 + LEFT*3), inf.animate.move_to(UP*2.55 + RIGHT*3))

        sinp = MathTex(r"\sin \pi z = \pi z \prod_{n=1}^\infty\left( 1 - \frac{z^2}{n^2} \right)")
        sinp.move_to(DOWN*0.5)
        sinpl = MathTex(r"L(\sin \pi z) = L(\pi z) + \sum_{n=1}^\infty L\left( 1 - \frac{z^2}{n^2} \right)")
        sinpl.move_to(DOWN*2.7)
        self.play(Write(sinp))
        self.wait()
        self.play(Write(sinpl))
        self.wait()
        sinpl2 = MathTex(r"\pi \cot \pi z = \frac{1}{z} + \sum_{n=1}^\infty \frac{2z}{z^2 - n^2}")
        sinpl2.move_to(DOWN*2.7)
        self.play(Transform(sinpl, sinpl2))
        self.wait()

        corr = MathTex(r"\frac{1}{z} - 2\frac{f(z)}{z} = \lim_{n\to\infty} \sum_{k=-n}^n \frac{1}{z-k} = \frac{1}{z} + \sum_{k=1}^\infty \frac{2z}{z^2-k^2}")
        corr.move_to(UP*3)
        self.play(FadeOut(inf, tit, sinp), FadeIn(corr))
        self.wait()
        corr2 = MathTex(r"\frac{1}{z} - 2\frac{f(z)}{z} = \pi \cot \pi z")
        self.play(Write(corr2))
        self.wait()

        corr3 = MathTex(r"-\frac{1}{2} + f(z) = -\frac{1}{2}\pi z \cot \pi z")
        self.play(Transform(corr2, corr3))
        self.wait()
        self.play(corr2.animate.shift(UP*3), FadeOut(sinpl, corr))
        zeta = MathTex(r"\zeta(s) = \sum_{n=1}^\infty \frac{1}{n^s}")
        zeta0 = Tex(r"By analytic continuation $\zeta(0) = -\frac{1}{2}$")
        #zeta.move_to(DOWN*2)
        zeta0.next_to(zeta, DOWN)
        self.play(Write(zeta))
        self.wait()
        self.play(Write(zeta0))
        self.wait()
        zetas = VGroup(zeta, zeta0)

        res1 = MathTex(r"\sum_{n=0}^\infty \zeta(2n)z^{2n} = -\frac{1}{2}\pi z \cot \pi z")
        res1.move_to(UP*0.5)
        self.play(zetas.animate.shift(DOWN*2))
        self.play(Write(res1))
        self.wait()
        self.play(FadeOut(zetas, corr2), res1.animate.shift(UP*2.5))
        self.wait()

        res2 = MathTex(r"\pi z\cot \pi z = \pi z \frac{\cos \pi z}{\sin \pi z} = \pi i z\frac{e^{\pi iz} + e^{-\pi iz}}{e^{\pi iz} - e^{-\pi iz}} = \pi i z + \frac{2\pi i z}{e^{2\pi iz}-1}")
        self.play(Write(res2))
        self.wait()
        res3 = MathTex(r"\pi z\cot \pi z = \frac{1}{2}\cdot 2\pi i z + \frac{2\pi i z}{e^{2\pi iz}-1}")
        self.play(Transform(res2, res3))
        self.wait()
        res4 = MathTex(r"-\frac{z}{2} - 2\sum_{n=0}^\infty \zeta(2n)\left(\frac{z}{2\pi i}\right)^{2n} =", r" \frac{z}{e^z - 1}")
        res4.move_to(DOWN*2.5)
        self.play(Write(res4))
        self.wait()
        self.play(FadeOut(res2, res1), res4.animate.move_to(UP*3))
        self.wait()

        bern1 = MathTex(r"\frac{z}{e^z - 1} = \sum_{n=0}^\infty \frac{B_n}{n!}z^n")
        bern2 = Tex(r"Bernoulli numbers $B_n$")
        bern1.move_to(UP*1)
        bern2.next_to(bern1, DOWN)
        self.play(FadeIn(bern1, bern2))
        self.wait()

        bern3 = MathTex(r"\left(\sum_{n=1}^\infty \frac{1}{n!}z^n\right)\left(\sum_{n=0}^\infty \frac{B_n}{n!} z^n\right) = z")
        bern3.move_to(DOWN*2)
        self.play(FadeIn(bern3))
        self.wait()
        self.play(FadeOut(res4, bern1), bern2.animate.move_to(UP*3), bern3.animate.move_to(UP*1.7))
        self.wait()
        bern4 = MathTex(r"B_0 = 1")
        bern5 = MathTex(r"\sum_{k=0}^{n-1} \frac{B_{k}}{k! (n-k)!} = 0 \quad (\text{if } n \ge 2)")
        bern6 = MathTex(r"\sum_{k=0}^{n-1} \binom{n}{k}B_k = 0 \quad (\text{if } n \ge 2)")
        bern4.move_to(UP*0.2)
        bern5.move_to(DOWN*1)
        bern6.move_to(DOWN*1)
        self.play(FadeIn(bern4))
        self.wait()
        self.play(FadeIn(bern5))
        self.wait()
        self.play(Transform(bern5, bern6))
        self.wait()
        ex1 = MathTex(r"B_0 + 2B_1 = 0 \Rightarrow B_1 = -\frac{1}{2}")
        ex2 = MathTex(r"B_0 + 3B_1 + 3B_2 = 0 \Rightarrow B_2 = \frac{1}{6}")
        ex3 = MathTex(r"B_0 + 4B_1 + 6B_2 + 4B_3 = 0 \Rightarrow B_3 = 0")
        ex4 = MathTex(r"B_0 + 5B_1 + 10B_2 + 10B_3 + 5B_4 = 0 \Rightarrow B_4 = -\frac{1}{30}")
        self.play(FadeOut(bern3), bern4.animate.move_to(UP*1.7 + LEFT*3), bern5.animate.move_to(UP*1.7 + RIGHT*3))
        ex2.next_to(ex1, DOWN)
        ex3.next_to(ex2, DOWN)
        ex4.next_to(ex3, DOWN)
        self.play(FadeIn(ex1, ex2, ex3, ex4))
        self.wait()
        self.play(FadeOut(ex1, ex2, ex3, ex4, bern4, bern5, bern2), FadeIn(res4))
        res4new = MathTex(r"\sum_{n=0}^\infty \frac{B_n}{n!} z^n")
        res4new.next_to(res4[0], RIGHT)
        self.wait()
        self.play(Transform(res4[1], res4new))
        self.wait()
        compare = MathTex(r"\zeta(2n) = \frac{(-1)^{n-1}(2\pi)^{2n}}{2 \cdot (2n)!} B_{2n}")
        compare2 = MathTex(r"\sum_{k=1}^\infty \frac{1}{k^{2n}} = \frac{(-1)^{n-1}(2\pi)^{2n}}{2 \cdot (2n)!} B_{2n}")
        self.play(Write(compare))
        self.wait()
        self.play(FadeOut(res4), Transform(compare, compare2))
        self.wait(2)
        self.play(FadeOut(compare))
        
```

```python
%%manim -qh S4

class S4(Scene):
    def construct(self):
        dirichlet = MathTex(r"\text{Dirichlet series: } \sum_{k=1}^\infty \frac{a_k}{k^s}")
        dirichlet.move_to(UP*2.5)
        eta = MathTex(r"\eta(s) = \sum_{n=1}^\infty \frac{(-1)^{n-1}}{n^s}")
        self.play(Write(dirichlet))
        self.wait()
        self.play(Write(eta))
        self.wait()
        self.play(FadeOut(dirichlet), eta.animate.shift(UP*3))

        sum1 = MathTex(r"\eta(s)", r" = 1 ", r" - \frac{1}{2^s}", r"+ \frac{1}{3^s}", r"- \frac{1}{4^s}", r"+ \frac{1}{5^s}", r"- \frac{1}{6^s}", r"+\cdots")
        sum2 = MathTex(r"\eta(s)", r" = 1 ", r" + \frac{1}{2^s}", r"+ \frac{1}{3^s}", r"+ \frac{1}{4^s}", r"+ \frac{1}{5^s}", r"+ \frac{1}{6^s}", r"+\cdots")
        sum3 = MathTex(r"\eta(s)", r" = 1 ", r" - \frac{2}{2^s}", r"+ \frac{1}{3^s}", r"- \frac{2}{4^s}", r"+ \frac{1}{5^s}", r"- \frac{2}{6^s}", r"-\cdots")
        sum4 = MathTex(r"\eta(s)", r" = (1-2^{1-s})\zeta(s)",  r" \frac{2}{4^s}", r"+ \frac{1}{5^s}", r"- \frac{2}{6^s}", r"-\cdots")
        sum1.move_to(UP*1.3)
        sum2.next_to(sum1, DOWN)
        sum3.next_to(sum2, DOWN)
        sum4.next_to(sum3, DOWN)
        self.play(FadeIn(sum1))
        self.wait()
        self.play(FadeIn(sum2[1:], sum3[2], sum3[4], sum3[6:8]))
        self.wait()
        self.play(Write(sum4[1]))
        self.wait()
        self.play(FadeOut(sum1, sum2[1:], sum3[2], sum3[4], sum3[6:8], sum4[1]))
        res = MathTex(r"\eta(s) = (1 - 2^{1-s})\zeta(s)")
        self.play(FadeIn(res))
        self.wait()
        self.play(eta.animate.shift(LEFT*3), res.animate.move_to(UP*3 + RIGHT*3))
        eval1 = MathTex(r"\sum_{k=1}^\infty (-1)^{k+1}\frac{1}{k^{2n}} = \eta(2n) = (1-2^{1-2n})\zeta(2n)")
        eval2 = MathTex(r"\sum_{k=1}^\infty \frac{1}{(2k-1)^{2n}} = \frac{1}{2}(\zeta(2n) + \eta(2n)) = \left(1 - \frac{1}{2^{2n}}\right) \zeta(2n)")
        eval1.move_to(UP*0.3)
        eval2.move_to(DOWN*2)
        self.play(FadeIn(eval1, eval2))
        self.wait()
        gen = Tex(r"Generating functions")
        gen.move_to(UP*3.5)

        self.play(FadeOut(eval1, eval2, eta, res))
        self.play(Write(gen))
        
        gen1 = MathTex(r"\sum_{n=0}^\infty \zeta(2n)z^{2n} = -\frac{\pi z}{2} \cot \pi z")
        gen2 = MathTex(r"\sum_{n=0}^\infty \frac{\zeta(2n) + \eta(2n)}{2}z^{2n} = \frac{\pi z}{4} \tan \frac{\pi z}{2}")
        gen3 = MathTex(r"\sum_{n=0}^\infty \eta(2n)z^{2n} = \frac{\pi z}{2}\frac{1}{\sin \pi z}")
        gen4 = MathTex(r"\sum_{n=0}^\infty \beta(2n+1)z^{2n+1} = \frac{\pi z}{4}\frac{1}{\cos \frac{\pi z}{2}}")
        gen1.move_to(UP*2.5)
        gen2.move_to(UP*0.55)
        gen3.move_to(DOWN*1.45)
        gen4.move_to(DOWN*3.3)
        self.play(FadeIn(gen1))
        self.wait()
        self.play(FadeIn(gen2, gen3))
        self.wait()
        self.play(Write(gen4))
        self.wait(2)
        self.play(FadeOut(gen, gen1, gen2, gen3), gen4.animate.move_to(UP*3))
        beta = MathTex(r"\beta(s) = \sum_{n=0}^\infty \frac{(-1)^n}{(2n+1)^s}")
        beta.move_to(UP*0.7)
        self.play(Write(beta))
        self.wait()
        eulnum = MathTex(r"\frac{1}{\cos z} = \sum_{n=0}^\infty \frac{(-1)^n E_{2n}}{(2n)!}z^{2n}")
        eulnum.move_to(DOWN*2)
        eulnum2 = Tex(r"Euler numbers $E_n$")
        eulnum2.next_to(eulnum, DOWN)
        self.play(Write(eulnum), Write(eulnum2))
        self.wait()
        eulnum3 = MathTex(r"E_0 = 1")
        eulnum4 = MathTex(r"\sum_{k=0}^n \binom{2n}{2k}E_{2k} = 0 \quad (\text{if } n \ge 1)")
        eulnum3.move_to(UP*0.7)
        eulnum4.move_to(DOWN*0.7)
        self.play(FadeOut(beta, gen4))
        self.play(VGroup(eulnum, eulnum2).animate.shift(UP*5))
        self.play(FadeIn(eulnum3, eulnum4))
        eulnum5 = MathTex(r"E_2 = -1, E_4 = 5, E_6 = -61, E_8 = 1385, \dots")
        eulnum5.move_to(DOWN*2.5)
        self.wait()
        self.play(FadeIn(eulnum5))
        self.wait()
        
        final = MathTex(r"\beta(2n+1) = \sum_{k=0}^\infty\frac{(-1)^{k}}{(2k+1)^{2n+1}}=\frac{(-1)^n E_{2n}\pi^{2n+1}}{4^{n+1}(2n)!}")
        final.move_to(DOWN*2.3)
        self.play(FadeOut(eulnum2, eulnum3, eulnum4, eulnum5), eulnum.animate.shift(DOWN*2.5), FadeIn(gen4))
        self.wait()
        self.play(Write(final))
        self.wait(3)
        self.play(FadeOut(gen4, eulnum), final.animate.move_to(UP*3))

        leibniz = MathTex(r"\frac{\pi}{4} = 1 - \frac{1}{3} + \frac{1}{5} - \frac{1}{7} + \frac{1}{9} - \frac{1}{11} +\cdots")
        leibniz2 = MathTex(r"\frac{\pi^3}{32} = 1 - \frac{1}{3^3} + \frac{1}{5^3} - \frac{1}{7^3} + \frac{1}{9^3} - \frac{1}{11^3} + \cdots")
        leibniz3 = MathTex(r"\frac{5\pi^5}{1536} = 1 - \frac{1}{3^5} + \frac{1}{5^5} - \frac{1}{7^5} + \frac{1}{9^5} - \frac{1}{11^5} + \cdots")
        leibniz.move_to(UP*0.7)
        leibniz2.next_to(leibniz, DOWN)
        leibniz3.next_to(leibniz2, DOWN)

        self.play(FadeIn(leibniz, leibniz2, leibniz3))
        self.wait()
        self.play(FadeOut(final, leibniz, leibniz2, leibniz3))
```

```python
%%manim -qh S5

class S5(Scene):
    def construct(self):
        tex = Tex(r"\textbf{Riemann zeta function} $\zeta(s)$")
        func = Tex(r"Functional equation \\$\zeta (s)=2^{s}\pi ^{s-1}\ \sin \left({\frac {\pi s}{2}}\right)\ \Gamma (1-s)\ \zeta (1-s)$")
        pos = MathTex(r"\zeta(2n) = \frac{(-1)^{n-1}(2\pi)^{2n}}{2 \cdot (2n)!} B_{2n}")
        neg = MathTex(r"\zeta(-n) = -\frac{B_{n+1}}{n+1}")
        tex.move_to(UP*3)
        func.move_to(UP*1)
        pos.move_to(DOWN*1.5)
        neg.next_to(pos, DOWN)

        self.play(Write(tex))
        self.wait()
        self.play(FadeIn(func))
        self.wait()
        self.play(FadeIn(pos))
        self.wait()
        self.play(Write(neg))
        self.wait()
        self.play(FadeOut(func, pos, neg))

        odd = Tex(r"What's the value of $\zeta(2n+1)$?")
        odd.move_to(UP*2)
        ape = MathTex(r"\zeta(3) = 1 + \frac{1}{2^3} + \frac{1}{3^3} + \frac{1}{4^3} + \frac{1}{5^3} + \frac{1}{6^3} + \cdots")
        ape2 = Tex(r"Apéry's constant")
        ape3 = Tex(r"proven to be irrational")
        ape.move_to(ORIGIN)
        ape2.next_to(ape, DOWN)
        ape3.next_to(ape2, DOWN)

        self.play(FadeIn(odd))
        self.wait()
        self.play(Write(ape))
        self.wait()
        self.play(Write(ape2))
        self.wait()
        self.play(FadeIn(ape3))
        self.wait()
        self.play(FadeOut(tex, odd, ape, ape2, ape3))
```

```python
%%manim -qh S6

class S6(Scene):
    def construct(self):
        tex = Tex(r"\textbf{Dirichlet beta function} $\beta(s)$")
        func = Tex(r"Functional equation \\$\beta (1-s)=\left({\frac {\pi }{2}}\right)^{-s}\sin \left({\frac {\pi }{2}}s\right)\Gamma (s)\beta (s)$")
        pos = MathTex(r"\beta (2n+1)\;=\;{\frac {(-1)^{n}E_{2n}}{2(2n)!}}\left({\frac {\pi }{2}}\right)^{2n+1}")
        neg = MathTex(r"\beta (-2n)\;=\;{\frac {1}{2}}E_{2n}")
        neg2 = MathTex(r"\beta (-2n-1)\;=\;0")
        tex.move_to(UP*3)
        func.move_to(UP*1.5)
        pos.move_to(DOWN*1)
        neg.next_to(pos, DOWN)
        neg2.next_to(neg, DOWN)

        self.play(Write(tex))
        self.wait()
        self.play(FadeIn(func))
        self.wait()
        self.play(FadeIn(pos))
        self.wait()
        self.play(Write(neg))
        self.play(Write(neg2))
        self.wait()
        self.play(FadeOut(tex, func, pos, neg, neg2))
```
