```python
%%manim -qh Recap

class Recap(Scene):
    def construct(self):
        last = ImageMobject("recap.png").scale(4/8)
        box = SurroundingRectangle(last, color=GRAY, buff=0)
        self.add(last, box)
        self.wait()
        im = Group(last, box)

        topo1 = Tex(r"Definition of a ", r"topological space", " (via ", r"open sets", r")", font_size=52).to_edge(UP)
        topo1[3].set_color(RED)
        topo1[1].set_color(BLUE)
        underline = Underline(topo1)
        toponeigh = Tex(r"A", r" topological space", r" $(X, \mathcal{T}_X)$ is a set $X$ with a collection \\ of subsets $\mathcal{T}_X \subset \mathcal{P}(X)$, called ", r"open sets", r", such that")
        toponeigh[1].set_color(BLUE)
        toponeigh[3].set_color(RED)
        toponeigh.next_to(underline, DOWN)
        
        naxioms = Tex(r"(i) $\emptyset$ and $X$ are open, \\", 
            r"(ii) if $(U_i)_{i\in I}$ are open, then $\bigcup_{i\in I} U_i$ is open, \\",
            r"(iii) if $U,V$ are open, then $U \cap V$ is open.")
        naxioms.arrange(DOWN, aligned_edge=LEFT, buff=0.25)
        naxioms[2:].shift(UP*0.1)
        naxioms.next_to(toponeigh, DOWN)

        top_group = VGroup(topo1, underline, toponeigh, naxioms)


        self.play(im.animate.shift(DOWN*2.4).scale(0.7), FadeIn(top_group))
        self.wait()
        self.play(FadeOut(im, top_group))

        
```

```python
%%manim -qh PartialOrdering1

class PartialOrdering1(Scene):
    def construct(self):
        title = Tex(r"Topologies ", r"on $X = \{a,b\}$").to_edge(UP)
        title[0].set_color(GREEN)
        underline = Underline(title)

        tp1 = MathTex(r"\{\emptyset, X\}").move_to(DOWN*2) # trivial topology
        tp2 = MathTex(r"\{\emptyset, \{a\}, X\}").move_to(LEFT*2.5)
        tp3 = MathTex(r"\{\emptyset, \{b\}, X\}").move_to(RIGHT*2.5)
        tp4 = MathTex(r"\{\emptyset, \{a\}, \{b\}, X\}").move_to(UP*2) # discrete topology

        topologies = VGroup(tp1, tp2, tp3, tp4)

        lines = VGroup(
            Arrow(end = UP*2, start=LEFT*2.5, buff=0.55), 
            Arrow(end=UP*2, start=RIGHT*2.5, buff=0.55), 
            Arrow(end=LEFT*2.5, start=DOWN*2, buff=0.55), 
            Arrow(end=RIGHT*2.5, start=DOWN*2, buff=0.55)
        )

        partial = Tex(r"partially ordered", r" by inclusion $\subseteq$").to_edge(DOWN)
        partial[0].set_color(BLUE)

        self.play(Write(title))
        self.play(Create(underline))
        self.wait()
        self.play(FadeIn(tp1, tp4))
        self.wait()
        self.play(FadeIn(tp2, tp3))
        self.wait()
        self.play(FadeIn(lines, partial))
        self.wait()

        order = VGroup(topologies, lines, partial)
        self.play(order.animate.to_edge(LEFT))

        finer = Tex(r"If $\mathcal{T}_1 \subseteq \mathcal{T}_2$, \\ then $\mathcal{T}_1$ is ", r"coarser ", r"than $\mathcal{T}_2$, \\ and $\mathcal{T}_2$ is ", r"finer", r" than $\mathcal{T}_1$.")
        finer[1].set_color(BLUE)
        finer[3].set_color(BLUE)
        finer.to_edge(RIGHT).shift(UP*1.5)

        self.play(FadeIn(finer))
        self.wait()

        self.play(FadeOut(topologies, title, underline, lines, partial, finer))

        # add: (maybe)
        # If f:X \to Y is continuous, then also with any finer topology on X and every coarser topology on Y
        
        
```

```python
%%manim -qh PartialOrdering2

class PartialOrdering2(Scene):
    def construct(self):
        topodef = Tex(r"Topologies $\mathcal{T}$ on $X$ such that $B \subseteq \mathcal{T}$", font_size=60).to_edge(UP)
        under = Underline(topodef)
        eg = Tex(r"e.g. $B = \{f^{-1}(U) \mid U \subseteq Y \text{ open}\}$").next_to(under, DOWN)

        self.play(FadeIn(topodef, under, eg))

        # Positions by layers: top (1), next (3), next (4), bottom (1)
        top = Dot(UP * 2).set_z_index(1)
        layer1 = VGroup(
            Dot(LEFT * 2 + UP * 0.5),
            Dot(LEFT * 0 + UP * 0.5),
            Dot(RIGHT * 2 + UP * 0.5),
        )
        layer2 = VGroup(
            Dot(LEFT * 3 + DOWN * 1),
            Dot(LEFT * 1 + DOWN * 1),
            Dot(RIGHT * 1 + DOWN * 1),
            Dot(RIGHT * 3 + DOWN * 1),
        )
        bottom = Dot(DOWN * 2.5).set_z_index(1)

        # Collect all dots so we can animate them together if wanted
        all_dots = VGroup(top, *layer1, *layer2, bottom)

        # Define edges (directed: from smaller -> larger)
        # Keep arrows short from dot edge to dot edge using buff
        def make_arrow(a: Mobject, b: Mobject, buff=0.12):
            return Arrow(
                a.get_center(),
                b.get_center(),
                buff=buff,
                tip_length=0.15,
                stroke_width=3,
                max_tip_length_to_length_ratio=0.2,
            )

        arrows = VGroup(
            make_arrow(bottom, layer2[2]),
            make_arrow(bottom, layer2[3]),
            # bottom -> some of layer2
            make_arrow(bottom, layer2[0]),
            make_arrow(bottom, layer2[1]),
            # connect layer2 among themselves to layer1 (partial order, not total)
            make_arrow(layer2[0], layer1[0]),
            make_arrow(layer2[1], layer1[0]),
            make_arrow(layer2[1], layer1[1]),
            make_arrow(layer2[2], layer1[1]),
            make_arrow(layer2[2], layer1[2]),
            make_arrow(layer2[3], layer1[2]),
            # some cross connections within layer2 -> layer1 to avoid chain collapse
            make_arrow(layer2[0], layer1[1]),
            # layer1 -> top (unique maximal)
            make_arrow(layer1[0], top),
            make_arrow(layer1[1], top),
            make_arrow(layer1[2], top),
        )


        # Optionally draw small subtle dots larger so they look like nodes
        for d in all_dots:
            d.set_radius(0.09)  # slightly larger dot for visibility

        # Center the whole graph nicely (optional)
        graph_group = VGroup(all_dots, arrows).arrange(
            IN, center=False
        )  # keep relative positions but group them
        graph_group.move_to(DOWN*0.75)

        # Draw: create dots first, then arrows
        self.play(
            AnimationGroup(
                *[Create(d) for d in all_dots],
                lag_ratio=0.06
            ),
            run_time=1.2
        )
        self.play(LaggedStart(*[Create(a) for a in arrows[2:]], lag_ratio=0.08), run_time=1.6)

        # Hold the final picture on screen briefly (so the scene doesn't immediately finish)
        self.wait()


        discretetop = Tex(r"Disrete Topology").set_color(BLUE).next_to(top, LEFT).shift(LEFT*0.25)
        self.play(top.animate.set_color(BLUE))
        self.play(FadeIn(discretetop))
        self.wait()

        self.play(VGroup(bottom, layer2[2], layer2[3]).animate.set_color(RED))
        self.wait()
        self.play(layer2[2:].animate.set_color(WHITE), FadeIn(arrows[0:2]))

        intertop = MathTex(r"\mathcal{T}_B", r" = \bigcap_{B\subseteq \mathcal{T}} \mathcal{T}")
        intertop.next_to(bottom, LEFT).shift(LEFT*0.85)
        intertop[0].set_color(RED)
        self.play(Write(intertop))
        
        self.wait()

        gentex = Tex(r"$\mathcal{T}_B$", r" is called \\ the topology generated by $B$.").next_to(bottom, RIGHT).shift(RIGHT*0.1)
        gentex[0].set_color(RED)
        self.play(FadeIn(gentex))
        self.wait()
        self.play(FadeOut(graph_group, topodef, under, eg, discretetop, intertop, gentex))
        
```

```python
%%manim -qh QuotientTopology

class QuotientTopology(Scene):
    def construct(self):        
        title = Tex(r"Quotient topology", font_size=55).to_edge(UP)
        underline = Underline(title)
        self.play(FadeIn(title, underline))
        
        tex1 = Tex(r"Given a topological space $Y$, \\ how to create a new space by ", r"gluing points together", r"?").next_to(underline, DOWN)
        tex1[1].set_color(GOLD)

        interval = NumberLine(x_range = [0,2,2], include_numbers=False, stroke_width=5.0).set_color(BLUE)
        arr = Arrow(start=LEFT*1.5, end=RIGHT*1.5)
        circle = Circle(radius=1)
        bsp1 = VGroup(interval, arr, circle).arrange(buff=1).next_to(tex1, DOWN, buff=0.75).shift(DOWN*0.2)
        bsp1tex = Tex(r"Identifying \\ end points", font_size=35).next_to(arr, UP, buff=0)
        bsp1.add(bsp1tex)
        
        self.play(Write(tex1))
        self.wait()
        self.play(FadeIn(bsp1))
        self.wait()
        self.play(FadeOut(bsp1))
        
        tex2 = Tex(r"Define a ", r"partition", r" $X$ on $Y$, so $X \subseteq \mathcal{P}(Y)$, $Y = \bigcup_{A \in X} A$, \\ $\emptyset \notin X$ and all sets in $X$ are pairwise disjoint.")
        tex2[1].set_color(RED)
        tex2.next_to(tex1, DOWN, buff=0.5)
        self.play(FadeIn(tex2))
        self.wait()

        grid = VGroup()
        for y in range(3):
            for x in range(3):
                dot = Dot(point=np.array([x - 1, 1 - y, 0]))
                grid.add(dot)

        grid.next_to(tex2, DOWN, buff=0.75)
        self.play(FadeIn(grid))
        self.wait(0.5)

        # Define groups for partition subsets
        A1_points = VGroup(*[grid[i] for i in range(6)])  # top two rows
        A2_points = VGroup(grid[6], grid[7])              # bottom-left two
        A3_points = VGroup(grid[8])                       # bottom-right one

        groups = [A1_points, A2_points, A3_points]
        colors = [RED, RED, RED]

        # Create rounded surrounding rectangles
        rectangles = [
            SurroundingRectangle(g, color=c, buff=0.25, corner_radius=0.15)
            for g, c in zip(groups, colors)
        ]

        # Animate the rectangles
        for rect in rectangles:
            self.play(Create(rect))

        self.wait()
        self.play(FadeOut(grid, *rectangles))
        
        
        def1 = Tex(r"The ", r"quotient topology", r" on $X$ is the", r" final topology", r" \\ induced by $q:Y \to X$, where $q$ is the unique map \\ such that $x \in q(x)$ for all $x\in Y$.")
        def1[1].set_color(GOLD)
        def1.shift(DOWN*2)
        self.play(Write(def1))
        self.wait()

        self.play(FadeOut(tex1, tex2))
        self.play(def1.animate.next_to(underline, DOWN), FadeIn(bsp1))
        self.wait()

        exp = Tex(r"$Y = [0,1]$, $X = \{\{0,1\}, \{x\} \mid x \in (0,1)\}$").move_to(DOWN*2)
        exp2 = Tex(r"$X$ with the ", r"quotient topology", r"\\ is topologically equivalent to a circle.")
        exp2.next_to(exp, DOWN)
        exp2[1].set_color(GOLD)
        self.play(FadeIn(exp))
        self.wait()
        self.play(Write(exp2))
        self.wait()
        self.play(FadeOut(def1, exp, exp2, bsp1))

        #equ1 = Tex(r"T", r"e", r"s", r"t")
        equ1 = Tex(r"The partition $X$ on $Y$ is most commonly given \\ by the set of all ", r"equivalence classes", r" of \\ an ", r"equivalence relation", r" $\sim$ on $Y$.")
        equ1[1].set_color(GREEN)
        equ1[3].set_color(GREEN)
        equ1.next_to(underline, DOWN)
        equ2 = MathTex(r"X = Y/\sim := \{[x]_\sim \mid x \in Y\}, \quad q:Y \to X, x \mapsto [x]_\sim").next_to(equ1, DOWN, buff=0.5)

        self.play(FadeIn(equ1, equ2))
        self.wait()

        equ3 = Tex(r"e.g. let $\sim$ be the ", r"equivalence relation", r" on $[0,1]$ \\ generated by $0 \sim 1$, then $[0,1]/{\sim } \cong S^1$.")
        equ3[1].set_color(GREEN)
        equ3.move_to(DOWN*1.5)
        self.play(Write(equ3))
        self.wait()

        self.play(FadeOut(equ1, equ2, equ3))

                # --- Square setup ---
        square = Square(side_length=2.5)
        square.move_to(LEFT * 4)
        square.set_stroke(width=3)
        square.set_fill(color=BLUE_E, opacity=0.3)

        # Corners
        p0 = square.get_corner(UL)
        p1 = square.get_corner(UR)
        p2 = square.get_corner(DR)
        p3 = square.get_corner(DL)

        # Helper: arrow along side, shifted *outward* from square
        def side_arrow(a, b, color, outward_shift=0.3):
            a, b = np.array(a), np.array(b)
            edge = b - a
            normal = np.array([edge[1], -edge[0], 0])  # outward normal
            normal /= np.linalg.norm(normal)
            shift_vec = normal * outward_shift
            return Arrow(
                start=a + shift_vec + 0.15 * edge,
                end=b + shift_vec - 0.15 * edge,
                buff=0,
                stroke_width=4,
                tip_length=0.18,
                color=color
            )

        # Colored arrows for each side (all oriented consistently)
        top_arrow = side_arrow(p0, p1, RED).shift(UP*0.55)
        bottom_arrow = side_arrow(p3, p2, RED)
        left_arrow = side_arrow(p3, p0, GREEN).shift(LEFT*0.55)
        right_arrow = side_arrow(p2, p1, GREEN)

        square_group = VGroup(square, top_arrow, bottom_arrow, left_arrow, right_arrow)

        # --- Middle arrow (center of scene) ---
        middle_arrow = Arrow(
            start=LEFT * 0.8,
            end=RIGHT * 0.8,
            buff=0,
            stroke_width=6,
            tip_length=0.25,
        )

        # --- Torus (rotated to show 3D) ---
        torus = Torus(major_radius=1.2, minor_radius=0.55)
        torus.move_to(RIGHT * 4)
        torus.set_fill(BLUE, opacity=0.6)
        torus.set_stroke(BLUE_E, width=0.5)
        torus.rotate(PI / 5, axis=RIGHT)
        torus.rotate(PI / 6, axis=UP)

        # --- Add everything ---

        equ = Tex(r"Let $\sim$ be the equivalence relation on $[0,1]^2$ generated by \\ $(t, 0) \sim (t,1) \, \forall t \in [0,1]$ and $(0, t) \sim (1,t) \, \forall t \in [0,1]$, \\ then $[0,1]^2 /{\sim } \cong T^2$.")
        equ.to_edge(DOWN).shift(DOWN*0.2)
        self.play(FadeIn(square_group, middle_arrow, torus, equ))
        self.wait()
        self.play(FadeOut(square_group, middle_arrow, torus, equ, title, underline))
        
```

```python
%%manim -qh CharacteristicProperty1

class CharacteristicProperty1(Scene):
    def construct(self):
        title = Tex(r"Characteristic property of the initial topology", font_size=55).to_edge(UP)
        under = Underline(title)
        
        setup = Tex(r"Suppose $X$ carries the initial topology \\ induced by $(f_i:X \to Y_i)_{i\in I}$.").next_to(under, DOWN)
        quest = Tex(r"When is ", r"$g:Z \to X$", r" continuous?").next_to(setup, DOWN)
        quest[1].set_color(GOLD)

        self.play(FadeIn(setup))
        self.play(Write(quest))
        self.wait()
        
        gen = Tex(r"Topology on $X$ is generated by $\bigcup_{i \in I} \{f_i^{-1}(U) \mid U \subseteq Y_i \text{ open}\}$")
        gen2 = Tex(r"$g:Z \to X$", r" is continuous if and only $g^{-1}(f_i^{-1}(U))$ is open \\ for every $i \in I$ and every open $U \subseteq Y_i$.")
        gen2[0].set_color(GOLD)
        
        gen.to_edge(DOWN)
        gen2.next_to(quest, DOWN)
        self.play(FadeIn(gen))
        self.wait()
        self.play(FadeIn(gen2))
        self.wait()
        

        res1 = Tex(r"$g^{-1}(f^{-1}_i(U)) = (f_i \circ g)^{-1}(U)$, where $f_i \circ g: Z \to Y_i$").next_to(gen2, DOWN)
        res2 = Tex(r"$g:Z \to X$", r" is continuous $\Leftrightarrow$ $f_i \circ g$ is continuous for every $i \in I$.").move_to(gen2)
        res2[0].set_color(GOLD)
        res3 = Tex(r"The initial topology is the ", r"unique", r" topology on $X$ \\ which satisfies this property.")
        res3.move_to(DOWN*2.75)
        res3[1].set_color(RED)
        
        self.play(FadeIn(res1), FadeOut(gen))
        self.wait()
        self.play(ReplacementTransform(gen2, res2))
        self.wait()
        self.play(Write(res3), FadeOut(res1))
        self.wait()
        self.play(FadeIn(title, under))
        self.wait()
        
        self.play(FadeOut(res2, res3, title, under, quest, setup))
```

```python
%%manim -qh CharacteristicProperty2

class CharacteristicProperty2(Scene):
    def construct(self):
        title1 = Tex(r"Characteristic property of the subspace topology", font_size=55).to_edge(UP)
        under1 = Underline(title1)
        sub1 = Tex(r"Suppose $A \subseteq X$ carries the subspace topology \\ (induced by the inclusion $i:A\hookrightarrow X$)").next_to(under1, DOWN)
        sub2 = Tex(r"$g:Z \to X$ is continuous $\Leftrightarrow$ $g\circ i : Z \to Y$ is continuous").next_to(sub1, DOWN)

        title2 = Tex(r"Characteristic property of the product topology", font_size=55).to_edge(UP).shift(DOWN*3.75)
        under2 = Underline(title2)
        prod1 = Tex(r"Suppose $X \times Y$ carries the product topology \\ (induced by $\text{pr}_X:X\times Y \twoheadrightarrow X$, $\text{pr}_Y:X\times Y \twoheadrightarrow Y$)").next_to(under2, DOWN)
        prod2 = Tex(r"$g=(g_X,g_Y): Z \to X \times Y$ is continuous $\Leftrightarrow$ $g_X = g \circ \text{pr}_X$ \\ and $g_Y = g \circ\text{pr}_Y$ are continuous.").next_to(prod1, DOWN)

        self.play(FadeIn(title1, under1, sub1, sub2))
        self.wait()
        self.play(FadeIn(title2, under2, prod1, prod2))
        self.wait()
        self.play(FadeOut(title1, under1, sub1, sub2, title2, under2, prod1, prod2))
        
```

```python
%%manim -qh CharacteristicProperty3

class CharacteristicProperty3(Scene):
    def construct(self):
        title = Tex(r"Characteristic property of the final topology").to_edge(UP)
        under = Underline(title)

        setup = Tex(r"Suppose $X$ carries the final topology \\ induced by $(f_i:Y_i \to X)_{i\in I}$").next_to(under, DOWN)
        
        res1 = Tex(r"$g:X \to Z$", r" is continuous $\Leftrightarrow$ $g \circ f_i$ is continuous for every $i \in I$.")
        res1[0].set_color(GOLD)
        res1.next_to(setup, DOWN).shift(DOWN*0.3)
        res2 = Tex(r"The final topology is the unique topology on $X$ \\ which satisfies this property.")
        res2.next_to(res1, DOWN).shift(DOWN*0.5)

        self.play(FadeIn(title, under, setup, res1, res2))
        self.wait()
        self.play(FadeOut(title, under, setup, res1, res2))
        
```

```python
%%manim -qh CharacteristicProperty4

class CharacteristicProperty4(Scene):
    def construct(self):
        title1 = Tex(r"Characteristic property of the disjoint union topology", font_size=55).to_edge(UP)
        under1 = Underline(title1)
        sub1 = Tex(r"Suppose $X \sqcup Y$ carries the disjoint union topology \\ (induced by $i_X:X\hookrightarrow X \sqcup Y$, $i_Y:Y\hookrightarrow X \sqcup Y$)").next_to(under1, DOWN)
        sub2 = Tex(r"$g = g_X \sqcup g_Y:X \sqcup Y \to Y$ is continuous $\Leftrightarrow$ $g_X = g \circ i_X$ \\ and $g_Y = g \circ i_Y$ are continuous.").next_to(sub1, DOWN)

        title2 = Tex(r"Characteristic property of the quotient topology", font_size=55).to_edge(UP).shift(DOWN*4)
        under2 = Underline(title2)
        prod1 = Tex(r"Suppose $X/{\sim }$ carries the quotient topology \\ (induced by $q:X \twoheadrightarrow X/{\sim }, x \mapsto [x]_\sim$)").next_to(under2, DOWN)
        prod2 = Tex(r"$g: X/{\sim } \to Z$ is continuous $\Leftrightarrow$ $g \circ q$ is continuous.").next_to(prod1, DOWN)
        prod3 = Tex(r"If $f: X \to Z$ is continuous and $f(x) = f(y)$ if $x \sim y$, then $\hat f:X/{\sim } \to Z, [x] \mapsto f(x)$ is well-defined and continuous.")
        ex1 = Tex(r"$f:[0,1] \to S^1, x \mapsto e^{2\pi i x}$ is continuous with $f(0) = f(1)$.")
        ex2 = Tex(r"$\Rightarrow$ $\hat f: [0,1] /_{0\sim 1} \to S^1$ is continuous.")

        self.play(FadeIn(title1, under1, sub1, sub2))
        self.wait()
        self.play(FadeIn(title2, under2, prod1, prod2))
        self.wait()
        self.play(FadeOut(title1, under1, sub1, sub2), VGroup(title2, under2, prod1, prod2).animate.to_edge(UP))
        prod3.next_to(prod2, DOWN, buff=0.75)
        ex1.next_to(prod3, DOWN, buff=0.75)
        ex2.next_to(ex1, DOWN)
        self.play(Write(prod3))
        self.wait()
        self.play(FadeIn(ex1, ex2))
        self.wait()
        self.play(FadeOut(title2, under2, prod1, prod2, prod3, ex1, ex2))
        
```
